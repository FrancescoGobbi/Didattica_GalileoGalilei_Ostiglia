```python
import array as arr

def sistema_array(v):
    secondo = v[1]
    penultimo = v[len(v) - 2]
    v[1] = penultimo
    v[len(v) - 2] = secondo

    for i in range(len(v)):
        if v[i] < 0:
            v[i] = -v[i]
        elif v[i] == 0:
            v[i] = i

def elabora_posizioni(v):
    for i in range(len(v)):
        if i % 2 == 0:
            v[i] = v[i] * 2
        else:
            v[i] = v[i] + 3

        if v[i] % 3 == 0:
            v[i] = v[i] - 1

if __name__ == "__main__":
    numeri = arr.array("i", [5, -4, 0, 7, -2, 6, 3])
    print("Array iniziale:", numeri)
    sistema_array(numeri)
    print("Array dopo sistema_array:", numeri)
    elabora_posizioni(numeri)
    print("Array finale:", numeri)
```
