```python
import array as arr

def cambia(v):
    temp = v[0]
    v[0] = v[len(v) - 1]
    v[len(v) - 1] = temp

    for i in range(len(v)):
        if v[i] < 0:
            v[i] = -v[i]
        elif i % 2 == 0:
            v[i] = v[i] + 2

def aggiorna(v):
    for i in range(len(v)):
        if v[i] % 2 == 0:
            v[i] = v[i] + i
        else:
            v[i] = v[i] * 2

if __name__ == "__main__":
    numeri = arr.array("i", [3, -5, 0, 4, -2])
    print(numeri)
    cambia(numeri)
    print(numeri)
    aggiorna(numeri)

    print(numeri)
```
