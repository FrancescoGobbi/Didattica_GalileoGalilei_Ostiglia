## 1. Che cos'è una funzione

Una **funzione** è un blocco di codice con un **nome**, che svolge un compito preciso e può essere **chiamato** più volte. Può **ricevere dati** e **restituire un risultato**, come $f(x) = 2x$.

Alcune funzioni sono già state usate: `print()`, `input()`, `len()`, `abs()`. Le parentesi indicano una **chiamata**; al loro interno si scrivono gli **argomenti**.

**Vantaggi.**

- **Riuso**: il codice si scrive una volta e si usa più volte.
- **Scomposizione**: un problema grande diventa un insieme di compiti piccoli.
- **Leggibilità**: il nome dice che cosa fa il blocco.
- **Correzione in un solo punto**: un errore si corregge una volta sola.

### `print()` produce un effetto, ma restituisce `None`

`print()` **mostra** un valore sullo schermo, ma l'oggetto che **restituisce** è `None`. L'effetto (la stampa) e il valore restituito sono **due aspetti distinti**.

**Output atteso:** `Ciao`, poi `None`.

````python
risultato = print("Ciao")  # Effetto: compare Ciao. Valore restituito: None.
print(risultato)  # Viene mostrato l'oggetto restituito da print.
````

### Perché scrivere una funzione: il riuso

Senza funzione, lo stesso calcolo va **riscritto** ogni volta. Con una funzione si scrive **una volta** e si chiama quante volte serve; se la formula cambia, si corregge in un solo punto.

**Output atteso:** `15`, `15`, `9.0`, `15`, `9.0`.

````python
# Senza funzione: la stessa formula viene scritta due volte.
print(3 * 5)  # Area del rettangolo 3 per 5.
print(2 * 4.5)  # Area del rettangolo 2 per 4.5: stessa formula, riscritta.
````

````python
def area_rettangolo(base, altezza):  # La formula viene scritta una sola volta.
    return base * altezza

print(area_rettangolo(3, 5))  # Prima chiamata.
print(area_rettangolo(2, 4.5))  # Seconda chiamata: nessuna formula da riscrivere.
````

## 2. Definire e chiamare una funzione

**Definire** significa scrivere l'intestazione `def` e il corpo: Python crea un **oggetto funzione** e lo associa a un'**etichetta**. Il corpo **non viene eseguito**.
**Chiamare** significa scrivere il nome seguito da parentesi tonde: solo ora il corpo viene eseguito, **una volta per ogni chiamata**.

Il modello è lo stesso delle lezioni precedenti: **etichetta → oggetto → classe**.

```text
saluta  ──►  oggetto funzione (codice del corpo)  ──►  classe function
```

### La definizione non esegue il corpo

Crea una funzione in python che saluta una persona. La cella seguente contiene **solo la definizione**: eseguendola **non compare nulla**, perché il corpo non viene eseguito.

**Output atteso:** nessuno.

````python
def saluta(nome):  # Definizione: nasce l'oggetto funzione, associato all'etichetta saluta.
    print("Ciao", nome)  # Corpo: non viene eseguito adesso.
````

### La funzione è un oggetto

L'etichetta `saluta` fa riferimento a un **oggetto**, istanza della classe `function`. Per questo si può mostrare con `print()` e se ne può chiedere la classe con `type()`.

**Output atteso:** `<class 'function'>`, poi una riga del tipo `<function saluta at 0x...>` (l'indirizzo cambia a ogni esecuzione).

````python
print(type(saluta))  # La classe dell'oggetto è function.
print(saluta)  # Viene mostrato l'oggetto funzione, non eseguito.
````

### Chiamare la funzione due volte

Ogni chiamata esegue **tutto il corpo** con l'argomento fornito. La stessa definizione serve per persone diverse.

**Output atteso:** `Ciao Anna`, `Ciao Luca`.

````python
saluta("Anna")  # Prima chiamata: nome riceve "Anna".
saluta("Luca")  # Seconda chiamata: nome riceve "Luca".
````

La versione seguente aggiunge stampe di controllo, per vedere **quando** il corpo viene eseguito: la riga `Fine definizione` compare **prima** di qualunque saluto.

**Output atteso:** `Fine definizione`, `Inizio chiamata`, `Ciao Marta`, `Fine chiamata`.

````python
def saluta(nome):  # Definizione: il corpo non viene eseguito.
    print("Ciao", nome)  # Corpo.

print("Fine definizione")  # Eseguita subito dopo il def.
print("Inizio chiamata")  # Solo ora arriva la chiamata.
saluta("Marta")  # Il corpo viene eseguito.
print("Fine chiamata")  # Si prosegue dopo il ritorno.
````

## 3. Struttura di una funzione in Python

```text
def  potenza(base, esponente):
         risultato = base ** esponente
         return risultato
```

| Parte | Ruolo |
|---|---|
| `def` | parola chiave che **definisce** la funzione |
| `potenza` | **nome** della funzione, con le stesse regole delle variabili |
| `(base, esponente)` | **parametri**: etichette che ricevono i dati |
| `:` | i **due punti** aprono il corpo |
| righe indentate | **corpo**, rientrato di quattro spazi |
| `return` | restituisce il valore e termina la funzione |

La **chiamata** ha la stessa forma di `print(12)` o `input("Numero: ")`: nome della funzione e argomenti tra parentesi. I parametri possono anche mancare: `def main():`.

### Una funzione costruita pezzo per pezzo

Prima la forma **essenziale**: la definizione e due chiamate, con il risultato mostrato da `print()`.

**Output atteso:** `8`, `25`.

````python
def potenza(base, esponente):  # Intestazione: nome e due parametri.
    risultato = base ** esponente  # Corpo: calcolo con un'etichetta locale.
    return risultato  # Il valore viene restituito a chi ha chiamato.

print(potenza(2, 3))  # 2 elevato a 3.
print(potenza(5, 2))  # 5 elevato a 2.
````

Poi la versione con **stampe di controllo**: dentro la funzione si vedono i dati ricevuti e il valore restituito.

**Output atteso:** `Ricevuti: base = 2 , esponente = 3`, `Restituisco: 8`, poi `Risultato: 8`.

````python
def potenza(base, esponente):  # Stessa funzione, con la traccia.
    print("Ricevuti: base =", base, ", esponente =", esponente)  # Dati ricevuti dai parametri.
    risultato = base ** esponente  # Calcolo.
    print("Restituisco:", risultato)  # Valore che sta per tornare al chiamante.
    return risultato

valore = potenza(2, 3)  # Il valore restituito viene associato a valore.
print("Risultato:", valore)  # Viene mostrato dopo il ritorno.
````

## 4. Parametro e argomento

| | Dove si trova | Che cos'è |
|---|---|---|
| **Parametro** | nella **definizione** | l'**etichetta** che esiste solo durante la chiamata e riceve l'oggetto dell'argomento |
| **Argomento** | nella **chiamata** | il **valore** passato: viene valutato **prima** e fornisce l'oggetto |

Gli argomenti vengono associati ai parametri **per posizione**: il primo argomento al primo parametro, il secondo al secondo.

### Parametri e argomenti nel codice

Crea una funzione in python che riceve una base e un esponente e restituisce la potenza. In `potenza(x, 3)` gli argomenti sono `x` (valutata prima: è l'intero 2) e `3`; i parametri sono `base` ed `esponente`.

**Output atteso:** `8`, `25`.

````python
def potenza(base, esponente):  # base ed esponente sono i PARAMETRI.
    return base ** esponente  # Il valore viene restituito.

def main():
    x = 2  # Etichetta di main: fa riferimento all'intero 2.
    print(potenza(x, 3))  # ARGOMENTI: x (valutata prima, vale 2) e 3.
    print(potenza(5, 2))  # ARGOMENTI: 5 e 2.

main()  # Chiamata di main.
````

La versione seguente mostra **che cosa riceve ciascun parametro** a ogni chiamata. Nella prima chiamata `base` riceve l'oggetto 2 e `esponente` l'oggetto 3; nella seconda `base` riceve 5 e `esponente` 2.

**Output atteso:** `base = 2 esponente = 3 -> 8`, `base = 5 esponente = 2 -> 25`.

````python
def potenza(base, esponente):  # Parametri.
    print("base =", base, "esponente =", esponente, end=" -> ")  # Che cosa è stato ricevuto.
    return base ** esponente  # Restituisce il risultato.

x = 2  # Un'etichetta usata come argomento.
print(potenza(x, 3))  # Argomenti: l'oggetto di x e il 3.
print(potenza(5, 2))  # Argomenti diversi, stessa funzione.
````

## 5. Il parametro è un'etichetta locale

Alla chiamata, il parametro viene **associato all'oggetto** passato come argomento, come in un'assegnazione: l'oggetto **non viene copiato**. Se nella funzione l'etichetta viene **riassegnata**, cambia solo l'etichetta **locale**.

Gli interi sono **immutabili**: `numero = numero + 1` calcola un **nuovo** intero e riassocia l'etichetta; l'intero 7 non cambia e l'etichetta `valore` di `main()` continua a farvi riferimento.

```text
Appena chiamata                       Dopo numero = numero + 1
valore ──► 7 : int                    valore ──► 7 : int
numero ──► 7 : int  (stesso)          numero ──► 8 : int
```

### Riassegnare un parametro

Crea una funzione in python che riceve un numero, lo aumenta di 1 e lo stampa.

**Output atteso:** `8`, poi `7`.

````python
def aumenta(numero):  # numero è un'etichetta locale.
    numero = numero + 1  # Calcola un nuovo intero e riassocia l'etichetta locale.
    print(numero)  # Viene mostrato 8.

def main():
    valore = 7  # Etichetta di main.
    aumenta(valore)  # numero riceve lo stesso oggetto di valore.
    print(valore)  # Ancora 7: l'intero non è cambiato.

main()
````

Con `id()` si vede se due etichette fanno riferimento allo **stesso oggetto**. All'inizio `numero` e `valore` indicano lo stesso intero; dopo la riassegnazione no.

**Output atteso:** `Stesso oggetto all'inizio: True`, `Stesso oggetto dopo l'aumento: False`, `valore vale ancora 7`.

````python
def aumenta(numero, originale):  # originale serve solo per il confronto.
    print("Stesso oggetto all'inizio:", id(numero) == id(originale))  # Stesso oggetto: True.
    numero = numero + 1  # Nuovo intero 8, associato all'etichetta locale.
    print("Stesso oggetto dopo l'aumento:", id(numero) == id(originale))  # Oggetti diversi: False.

valore = 7  # Etichetta esterna alla funzione.
aumenta(valore, valore)  # Entrambi i parametri fanno riferimento all'oggetto 7.
print("valore vale ancora", valore)  # L'oggetto 7 non è stato modificato.
````

### Le etichette locali esistono solo durante la chiamata

Crea una funzione in python che riceve due voti e restituisce la loro media. L'etichetta `somma` è **locale**: nasce con la chiamata e **non esiste più** quando la funzione termina. All'esterno arriva solo il **valore restituito**.

## 6. `return`: restituire non è stampare

- `return` **termina** la funzione e consegna un oggetto al **chiamante**, che può associarlo a un'etichetta o usarlo in un'espressione.
- Una funzione senza `return` restituisce comunque `None`, l'oggetto che rappresenta **nessun valore**.
- `print()` mostra un valore, ma **non** lo rende disponibile al resto del programma: un valore da usare nei calcoli va **restituito** con `return`.
- Le istruzioni scritte **dopo** un `return` eseguito non vengono raggiunte.

### Una funzione che restituisce e una che stampa

Crea una funzione in python che restituisce il doppio di un numero e una che lo stampa.

**Output atteso:** `7`, poi `6 None`.

````python
def doppio(x):
    return 2 * x  # Restituisce l'oggetto al chiamante.

def mostra(x):
    print(x)  # Mostra soltanto: la funzione termina restituendo None.

def main():
    r = doppio(3)  # r fa riferimento a 6.
    e = mostra(7)  # Compare 7; e fa riferimento a None.
    print(r, e)  # Viene mostrato 6 None.

main()
````

Il valore **restituito** si può riusare in un calcolo; quello **solo stampato** no. La cella seguente mostra la differenza.

**Output atteso:** `12`, `8`, `8` (la seconda stampa viene dalla chiamata dentro `try`), poi `Errore: unsupported operand type(s) for +: 'NoneType' and 'int'`.

````python
def doppio(x):
    return 2 * x  # Restituisce 2 * x.

def mostra(x):
    print(2 * x)  # Stampa 2 * x.

print(doppio(4) + doppio(2))  # Si possono sommare i due valori restituiti: 8 + 4.
mostra(4)  # Compare 8, ma il valore non è disponibile al chiamante.

try:  # La prova viene protetta per mostrare l'errore senza interrompere la cella.
    totale = mostra(4) + 1  # mostra restituisce None: None + 1 non è possibile.
except TypeError as errore:  # Viene catturato il TypeError.
    print("Errore:", errore)  # Viene mostrato il messaggio di Python.
````

## 7. Il modulo e la funzione `main()`

Un **modulo** è un file `.py`. Il codice fuori dalle funzioni è il **codice del modulo**. Un programma ben ordinato ha tre zone:

1. **Definizioni**: le istruzioni `def` creano le funzioni, **senza eseguirne il corpo**.
2. **Controllo**: `__name__ == '__main__'` è vero se il file è **eseguito direttamente**.
3. **Avvio**: la chiamata `main()` esegue il corpo della funzione principale.

Python legge il modulo **dall'alto verso il basso**. Il programma vero e proprio parte solo all'ultima chiamata.

### Un programma con `main()`

Crea una funzione in python che stampa un saluto e avvia il programma dalla funzione `main()`.

**Output atteso:** `Ciao!`.

````python
def saluta():  # Definizione: crea l'oggetto funzione.
    print("Ciao!")  # Corpo: fermo fino alla chiamata.

def main():  # Programma principale.
    saluta()  # Chiamata a saluta.

if __name__ == '__main__':  # Controllo: vero se il file è eseguito direttamente.
    main()  # Avvio del programma.
````

### L'ordine di esecuzione

Le stampe numerate mostrano **in che ordine** Python esegue le righe. Prima vengono lette le due definizioni, poi il controllo e solo alla fine parte `main()`.

**Output atteso:** `1. Il modulo viene letto`, `2. Il controllo di __name__`, `3. Parte main()`, `4. Dentro saluta()`, `5. Fine di main()`.

````python
print("1. Il modulo viene letto")  # Prima riga del modulo.

def saluta():  # Definizione: il corpo non viene eseguito adesso.
    print("4. Dentro saluta()")  # Eseguita solo quando saluta viene chiamata.

def main():  # Definizione: il corpo non viene eseguito adesso.
    print("3. Parte main()")  # Prima istruzione del corpo di main.
    saluta()  # Chiamata: il controllo passa a saluta.
    print("5. Fine di main()")  # Eseguita dopo il ritorno da saluta.

print("2. Il controllo di __name__")  # Si arriva qui dopo i def.
if __name__ == '__main__':  # Vero: il file è eseguito direttamente.
    main()  # Avvio.
````

## 8. Funzioni che chiamano funzioni

Una funzione può chiamarne altre. Mentre la funzione chiamata è in esecuzione, quella chiamante **attende**.

### Seguire le chiamate nel codice

Crea una funzione in python che calcola il quadrato e una che somma i quadrati di due numeri.

Forma **essenziale**: il programma completo, con `main()`.

**Output atteso:** `25`.

````python
def quadrato(x):  # Riceve un numero e restituisce il suo quadrato.
    return x * x

def somma_quadrati(a, b):  # Chiama quadrato due volte.
    return quadrato(a) + quadrato(b)

def main():
    risultato = somma_quadrati(3, 4)  # Calcola 3 al quadrato più 4 al quadrato.
    print(risultato)  # Mostra 25.

if __name__ == '__main__':
    main()
````

### Vedere lo stack con una funzione di controllo

La funzione `pila()` elenca i frame attivi **dalla base alla cima**, ricavandoli dalle chiamate in corso. Serve solo a osservare lo stack: nei programmi normali non si usa. Il modulo viene aggiunto a mano come base.

La stessa lezione, con una stampa di controllo **all'inizio** e **alla fine** di ogni funzione.

**Output atteso:** si vede la pila crescere con le chiamate e ridursi con i ritorni; l'ultima riga è `Risultato: 25`.

````python
import traceback  # Modulo che permette di leggere le chiamate in corso.

def pila():  # Funzione di controllo: elenca i frame delle sole funzioni della lezione.
    nomi = [riga.name for riga in traceback.extract_stack()[:-1]]  # Nomi delle funzioni in corso.
    return ["modulo"] + [n for n in nomi if n in ("main", "somma_quadrati", "quadrato")]

def quadrato(x):
    print("  entra quadrato(", x, ") pila:", pila())  # Frame appena aggiunto in cima.
    r = x * x  # Calcolo locale.
    print("  esce  quadrato(", x, ") restituisce", r)  # Sta per essere tolto.
    return r

def somma_quadrati(a, b):
    print(" entra somma_quadrati(", a, ",", b, ") pila:", pila())  # Frame aggiunto.
    totale = quadrato(a) + quadrato(b)  # Due chiamate una dopo l'altra.
    print(" esce  somma_quadrati restituisce", totale)  # Sta per essere tolto.
    return totale

def main():
    print("entra main() pila:", pila())  # Frame di main appena aggiunto.
    risultato = somma_quadrati(3, 4)  # Chiamata.
    print("Risultato:", risultato)  # Dopo il ritorno, main riprende.

main()  # Il modulo chiama main().
````

## 9. Esempio

**Funzione che calcola il prezzo scontato**

Crea una funzione in python che riceve il **prezzo** di un prodotto e la **percentuale di sconto** e restituisce il **prezzo scontato**.

````python
def prezzo_scontato(prezzo, percentuale):  # Intestazione: due parametri.
    sconto = prezzo * percentuale / 100  # sconto è locale: vale 20.0 nella prima chiamata.
    return prezzo - sconto  # Restituisce il prezzo scontato.

def main():
    finale = prezzo_scontato(80, 25)  # prezzo riceve 80, percentuale riceve 25.
    print(finale)  # Viene mostrato 60.0.
    print(prezzo_scontato(50, 10))  # Seconda prova.
    print(prezzo_scontato(20, 0))  # Con sconto 0 il prezzo non cambia.

if __name__ == '__main__':
    main()
````

Con le **stampe di controllo** si vedono i dati ricevuti, l'etichetta locale `sconto` e il valore restituito. `/` produce sempre un `float`, quindi il risultato è `60.0` e non `60`.

**Output atteso:** `prezzo = 80 percentuale = 25`, `sconto (locale) = 20.0`, `restituisce 60.0`, poi `Prezzo finale: 60.0`.

````python
def prezzo_scontato(prezzo, percentuale):
    print("prezzo =", prezzo, "percentuale =", percentuale)  # Dati ricevuti dai parametri.
    sconto = prezzo * percentuale / 100  # Etichetta locale.
    print("sconto (locale) =", sconto)  # Lo sconto calcolato.
    nuovo = prezzo - sconto  # Prezzo scontato.
    print("restituisce", nuovo)  # Valore che torna al chiamante.
    return nuovo

finale = prezzo_scontato(80, 25)  # Prima chiamata di prova.
print("Prezzo finale:", finale)  # Stampa spetta a chi chiama.
````

### Media di tre voti e giudizio

**Consegna.** Crea una funzione in python che riceve **tre voti** e **restituisce** la loro **media**. Crea poi una funzione in python che riceve la media e **restituisce** `"sufficiente"` se è almeno 6, altrimenti `"insufficiente"`. La `main()` stampa media e giudizio; le due funzioni **non** contengono `print()`.

**Soluzione spiegata.** La seconda funzione usa il valore restituito dalla prima: la `main()` chiama `media`, associa il risultato a `m` e lo passa a `giudizio`. Con 6, 7 e 8 la media è `7.0`; con 5, 6 e 4 è `5.0`.

**Output atteso:** `7.0 sufficiente`, `5.0 insufficiente`.

````python
def media(a, b, c):  # Tre parametri, tre voti.
    return (a + b + c) / 3  # Restituisce la media (un float).

def giudizio(m):  # Riceve la media.
    if m >= 6:  # Caso sufficiente.
        return "sufficiente"
    else:  # Caso insufficiente.
        return "insufficiente"

def main():
    m = media(6, 7, 8)  # Il risultato di media diventa l'argomento di giudizio.
    print(m, giudizio(m))  # Stampa media e giudizio.
    m = media(5, 6, 4)  # Seconda prova.
    print(m, giudizio(m))  # Stampa media e giudizio.

if __name__ == '__main__':
    main()
````
