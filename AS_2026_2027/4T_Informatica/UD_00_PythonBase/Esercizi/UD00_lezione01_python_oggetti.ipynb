## La classe Automobile

### Concetto
Una **classe** definisce un tipo di oggetti: descrive le caratteristiche e i comportamenti comuni alle sue istanze. Una **istanza**, o oggetto, è un elemento concreto di quel tipo, con una propria identità e un proprio stato.

Gli **attributi** descrivono lo stato; i **metodi** descrivono le operazioni. `Automobile` prevede gli attributi `colore` e `velocita` e il metodo `accelera`.

`class` introduce la classe. `def` definisce un metodo. `self` fa riferimento all’istanza su cui il metodo opera. `__init__` inizializza un’istanza appena creata: in questo esempio imposta colore e velocità. L’indentazione raggruppa le istruzioni. `+=` aggiunge un valore e riassegna il risultato all’attributo.

### Passaggi del codice
1. Definire la classe `Automobile`, senza creare ancora istanze.
2. In `__init__`, associare a ciascuna istanza il colore ricevuto e la velocità iniziale 0.
3. In `accelera`, aggiornare la velocità dell’istanza tramite `self.velocita += incremento`.

### Cosa osservare
La cella non stampa nulla. Definire un metodo non significa eseguirlo: il suo corpo verrà eseguito quando il metodo sarà chiamato. `self` permette a istanze diverse di usare gli stessi metodi mantenendo stati distinti.


```python
class Automobile:  # Definisce il modello comune delle automobili.
    def __init__(self, colore):  # Inizializza lo stato della nuova istanza.
        self.colore = colore  # Associa l’attributo dell’istanza al valore iniziale.
        self.velocita = 0  # Associa l’attributo dell’istanza al valore iniziale.

    def accelera(self, incremento):  # Definisce il metodo che aumenta la velocità.
        self.velocita += incremento  # Aggiorna l’attributo velocità aggiungendo l’incremento.
```

## Due etichette e un risultato

### Concetto
Con `b = a`, le due variabili fanno riferimento allo stesso oggetto: questa situazione si chiama **aliasing**. Dopo `a = risultato`, soltanto `a` cambia riferimento.

Prima: **a, b → int 16**. Dopo: **a, risultato → int 17**; **b → int 16**.

### Passaggi del codice
1. Associare `a` e `b` allo stesso intero 16.
2. Calcolare 16 + 1 e associare `risultato` all’intero ottenuto.
3. Riassegnare soltanto `a`, poi stampare `a` e `b`.

### Cosa osservare
L’output è `17` e `16`. `b = a` condivide l’oggetto di quel momento, ma non crea un collegamento che obblighi `b` a seguire tutte le future riassegnazioni di `a`.


```python
a = 16  # Associa la variabile all’oggetto indicato.
b = a  # Associa un’altra etichetta allo stesso oggetto.

risultato = a + 1  # Valuta l’espressione e associa la variabile al risultato.
a = risultato  # Riassegna a al risultato; b mantiene il riferimento precedente.

print(a)  # Mostra il valore o il risultato indicato.
print(b)  # Mostra il valore o il risultato indicato.
```

## 9. Creare una classe personalizzata: Contatore

### Consegna
Definire una classe `Contatore` con un valore iniziale e un metodo che lo aumenti. Creare due istanze indipendenti, modificare la prima tramite un’altra etichetta e visualizzare lo stato di entrambe.

### Concetto
Una **classe** definisce un tipo di oggetti e i comportamenti comuni alle sue istanze. Un’**istanza** è un oggetto concreto di quella classe. Gli **attributi di istanza** ne rappresentano lo stato; i **metodi** operano sull’istanza.

`class` introduce la classe; `def` introduce un metodo. `__init__` inizializza lo stato dell’istanza appena creata. `self` riceve il riferimento all’istanza su cui viene chiamato il metodo. Nell’espressione `self.valore`, il punto permette di accedere al suo attributo.

### Passaggi della soluzione
1. Inizializzare l’attributo `valore` con il parametro ricevuto.
2. Definire `aumenta()` per aggiornare l’attributo.
3. Creare due oggetti distinti.
4. Associare `alias` al primo oggetto e chiamare il metodo tramite questa etichetta.

### Risultato atteso
Il primo contatore passa da 0 a 3; il secondo rimane a 10. `primo is alias` vale `True`, mentre `primo is secondo` vale `False`.

**Schema:** `primo → oggetto Contatore con valore 3 ← alias`; `secondo → altro oggetto Contatore con valore 10`.

```python
class Contatore:  # Definisce la classe degli oggetti contatore.
    def __init__(self, iniziale):  # Riceve l’istanza e il riferimento al valore iniziale.
        self.valore = iniziale  # Inizializza l’attributo dell’istanza con il dato ricevuto.

    def aumenta(self, incremento):  # Definisce il metodo che aggiorna il contatore.
        self.valore += incremento  # Calcola la somma e aggiorna l’attributo dell’istanza.

primo = Contatore(0)  # Crea un’istanza con valore iniziale zero.
secondo = Contatore(10)  # Crea un’istanza distinta con valore iniziale dieci.
alias = primo  # Associa una seconda etichetta allo stesso oggetto di primo.
alias.aumenta(3)  # Modifica il primo oggetto tramite l’etichetta alias.
print(primo.valore)  # Mostra 3: la modifica è visibile anche tramite primo.
print(secondo.valore)  # Mostra 10: il secondo oggetto non è stato modificato.
print(primo is alias)  # Mostra True: le due etichette indicano lo stesso oggetto.
print(primo is secondo)  # Mostra False: le due istanze sono distinte.
```
