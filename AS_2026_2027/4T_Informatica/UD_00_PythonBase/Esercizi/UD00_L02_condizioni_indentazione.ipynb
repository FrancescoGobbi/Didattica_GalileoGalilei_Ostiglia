## Concetti chiave
- Una **variabile** è un’**etichetta** che fa riferimento a un oggetto. “Punta” è una rappresentazione concettuale, non necessariamente un puntatore RAM come in C.
- `=` esegue un’**assegnazione**; `==` confronta due valori.
- Una **condizione** è un’espressione di cui Python verifica il valore di verità. I confronti numerici qui usati producono `True` o `False`, oggetti della classe `bool`.
- Un **blocco** raggruppa istruzioni mediante l’**indentazione**. Si usano quattro spazi per livello. Il ritorno al livello esterno chiude il blocco.
- `if` introduce la prima condizione; `elif` introduce una condizione alternativa; `else` gestisce il caso residuo senza una condizione propria. Le intestazioni terminano con `:`.
- `and` richiede entrambe le condizioni, `or` almeno una, `not` inverte il valore di verità. Con operandi booleani i risultati sono booleani; in generale `and` e `or` restituiscono uno degli operandi.



## Confronti e operatori logici

| Operatore | Significato |
|---|---|
| `==`, `!=` | uguale a, diverso da |
| `<`, `>` | minore di, maggiore di |
| `<=`, `>=` | minore o uguale a, maggiore o uguale a |

| A | B | A and B | A or B |
|---|---|---|---|
| `False` | `False` | `False` | `False` |
| `False` | `True` | `False` | `True` |
| `True` | `False` | `False` | `True` |
| `True` | `True` | `True` | `True` |

`not True` vale `False`; `not False` vale `True`. Tra questi operatori, i confronti hanno precedenza su `not`, poi vengono `and` e infine `or`. Le parentesi rendono esplicito il raggruppamento: `True or False and False` vale `True`, mentre `(True or False) and False` vale `False`.

## Blocchi ed errori

Le intestazioni `if`, `elif` ed `else` terminano con `:`. Ogni blocco deve contenere almeno un’istruzione; una riga vuota o un commento non bastano. `pass` è un’istruzione che non compie azioni e può occupare provvisoriamente un blocco.

Quattro spazi sono la convenzione adottata: la sintassi richiede livelli coerenti. Non mescolare spazi e tabulazioni. `IndentationError` segnala un problema nella struttura dei rientri; `TabError` ne è un caso particolare. Un **errore logico**, invece, può produrre un risultato sbagliato anche quando la sintassi è corretta.

## 1. Correggere i blocchi

**Problema**

`if` ed `else` sono allineati; ogni stampa del ramo ha quattro spazi. La stampa finale è esterna. Con 7 si ottengono “Promosso” e “Fine verifica”.

```python
voto = 7  # Associa voto al risultato dell’espressione a destra.
if voto >= 6:  # Verifica la condizione per decidere se eseguire il blocco.
    print("Promosso")  # Mostra il valore indicato quando questa istruzione viene raggiunta.
else:  # Introduce il ramo eseguito quando le condizioni precedenti sono false.
    print("Da recuperare")  # Mostra il valore indicato quando questa istruzione viene raggiunta.
print("Fine verifica")  # Mostra il valore indicato quando questa istruzione viene raggiunta.
```

## 3. Controllare un accesso

**Problema**

Consentire l’accesso soltanto con autorizzazione e badge. In caso di rifiuto distinguere il requisito mancante.

**Procedimento**

La selezione esterna verifica il permesso. Quella interna verifica il badge. Con entrambi i valori veri si ottiene “Entra”.

```python
autorizzato = True  # Associa autorizzato al risultato dell’espressione a destra.
badge = True  # Associa badge al risultato dell’espressione a destra.
if autorizzato == True:  # Verifica la condizione per decidere se eseguire il blocco.
    if badge == True:  # Verifica la condizione per decidere se eseguire il blocco.
        print("Entra")  # Mostra il valore indicato quando questa istruzione viene raggiunta.
    else:  # Introduce il ramo eseguito quando le condizioni precedenti sono false.
        print("Manca il badge")  # Mostra il valore indicato quando questa istruzione viene raggiunta.
else:  # Introduce il ramo eseguito quando le condizioni precedenti sono false.
    print("Manca il permesso")  # Mostra il valore indicato quando questa istruzione viene raggiunta.
```

## 4. Eseguire verifiche indipendenti

**Problema**

Dato un intero, indicare se è positivo e se è pari. I messaggi devono poter comparire insieme. Stampare sempre “Verifica terminata”.

**Procedimento**

Le proprietà non sono alternative: un numero può essere positivo e pari. Si utilizzano due `if` indipendenti. Il secondo controllo viene eseguito anche se il primo è vero.

**Risultato atteso**

Con 8 compaiono “Positivo”, “Pari” e “Verifica terminata”. Con -3 compare soltanto il messaggio finale.

```python
numero = 8  # Imposta l’intero da esaminare.
if numero > 0:  # Verifica la positività del numero.
    print("Positivo")  # Mostra il messaggio soltanto per valori maggiori di zero.
if numero % 2 == 0:  # Verifica la parità indipendentemente dal primo controllo.
    print("Pari")  # Mostra il messaggio se il resto della divisione per 2 è zero.
print("Verifica terminata")  # Mostra il messaggio dopo entrambe le selezioni.
```

## 5. Validare e classificare un punteggio

**Problema**

Un punteggio deve essere compreso tra 0 e 100. Mostrare “Non valido” fuori da questo intervallo, “Livello alto” da 80 a 100, “Livello base” da 60 incluso a 80 escluso e “Da migliorare” da 0 incluso a 60 escluso.

**Procedimento**

La prima condizione esclude i dati fuori intervallo usando `or`. La catena prosegue soltanto per valori validi. Le soglie sono controllate dalla più alta alla più bassa. Ogni ramo assegna `esito`, che viene stampato dopo la scelta.

**Risultato atteso**

Con 75 compare “Livello base”. I valori -1, 0, 59, 60, 79, 80, 100 e 101 permettono di controllare tutti i rami e i confini.

```python
punteggio = 75  # Imposta il punteggio da verificare.
if punteggio < 0 or punteggio > 100:  # Rileva un valore fuori dall’intervallo ammesso.
    esito = "Non valido"  # Associa esito al messaggio per i dati non validi.
elif punteggio >= 80:  # Controlla la soglia più alta tra i punteggi validi.
    esito = "Livello alto"  # Classifica i punteggi da 80 a 100.
elif punteggio >= 60:  # Controlla la soglia successiva dopo aver escluso la fascia alta.
    esito = "Livello base"  # Classifica i punteggi da 60 incluso a 80 escluso.
else:  # Gestisce i punteggi validi inferiori a 60.
    esito = "Da migliorare"  # Assegna il messaggio per la fascia inferiore.
print(esito)  # Mostra l’unico esito scelto dalla catena.
```
