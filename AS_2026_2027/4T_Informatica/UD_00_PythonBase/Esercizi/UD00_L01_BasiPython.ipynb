## La classe Automobile
Una **classe** definisce un tipo di oggetti: descrive le caratteristiche e i comportamenti comuni alle sue istanze. Una **istanza**, o oggetto, è un elemento concreto di quel tipo, con una propria identità e un proprio stato.

Gli **attributi** descrivono lo stato; i **metodi** descrivono le operazioni. `Automobile` prevede gli attributi `colore` e `velocita` e il metodo `accelera`.

`class` introduce la classe. `def` definisce un metodo. `self` fa riferimento all’istanza su cui il metodo opera. `__init__` inizializza un’istanza appena creata: in questo esempio imposta colore e velocità. L’indentazione raggruppa le istruzioni. `+=` aggiunge un valore e riassegna il risultato all’attributo.

```python
class Automobile:  # Definisce il modello comune delle automobili.
    def __init__(self, colore):  # Inizializza lo stato della nuova istanza.
        self.colore = colore  # Associa l’attributo dell’istanza al valore iniziale.
        self.velocita = 0  # Associa l’attributo dell’istanza al valore iniziale.

    def accelera(self, incremento):  # Definisce il metodo che aumenta la velocità.
        self.velocita += incremento  # Aggiorna l’attributo velocità aggiungendo l’incremento.
```

## Creare un oggetto
La chiamata `Automobile("blu")` crea un’istanza e ne inizializza lo stato. La variabile `auto` è un’etichetta che fa riferimento all’oggetto. Il punto consente di accedere a un attributo.

**auto → istanza con colore blu e velocità 0 → classe Automobile**

`type(auto) is Automobile` verifica se il tipo dell’oggetto è proprio la classe `Automobile`.

```python
# Dopo la definizione della classe
auto = Automobile("blu")  # Crea un’istanza e associa la variabile all’oggetto.
print(auto.colore)  # Mostra il valore o il risultato indicato.
print(auto.velocita)  # Mostra il valore o il risultato indicato.
print(type(auto) is Automobile)  # Mostra il tipo dell’oggetto.
```

## Stato e metodo
La chiamata `auto.accelera(20)` esegue il metodo sull’istanza indicata da `auto`. La velocità passa da 0 a 20: cambia lo stato dello stesso oggetto.

```python
auto = Automobile("blu")  # Crea un’istanza e associa la variabile all’oggetto.
print(auto.velocita)  # Mostra il valore o il risultato indicato.
auto.accelera(20)  # Chiama il metodo sull’istanza indicata dalla variabile.
print(auto.velocita)  # Mostra il valore o il risultato indicato.
```

## Due istanze distinte
Due chiamate a `Automobile` creano due oggetti distinti. La modifica di `altra` non cambia lo stato di `auto`. `is` confronta l’identità, non il valore.

**auto → automobile blu, velocità 0**  
**altra → automobile rossa, velocità 30**

```python
auto = Automobile("blu")  # Crea un’istanza e associa la variabile all’oggetto.
altra = Automobile("rossa")  # Crea un’istanza e associa la variabile all’oggetto.
altra.accelera(30)  # Chiama il metodo sull’istanza indicata dalla variabile.
print(auto.colore, auto.velocita)  # Mostra il valore o il risultato indicato.
print(altra.colore, altra.velocita)  # Mostra il valore o il risultato indicato.
print(type(auto) is type(altra))  # Mostra il tipo dell’oggetto.
print(auto is altra)  # Confronta l’identità degli oggetti.
```

## Riassegnare una variabile
La riassegnazione cambia l’oggetto a cui una variabile fa riferimento. L’intero 16 non si trasforma in una stringa: la variabile `x` viene associata a un altro oggetto.

```python
x = 16  # Associa la variabile all’oggetto indicato.

print(x)  # Mostra il valore o il risultato indicato.
print(type(x))  # Mostra il tipo dell’oggetto.

x = "sedici"  # Associa la variabile all’oggetto indicato.

print(x)  # Mostra il valore o il risultato indicato.
print(type(x))  # Mostra il tipo dell’oggetto.
```

## Due etichette e un risultato
Con `b = a`, le due variabili fanno riferimento allo stesso oggetto: questa situazione si chiama **aliasing**. Dopo `a = risultato`, soltanto `a` cambia riferimento.

Prima: **a, b → int 16**. Dopo: **a, risultato → int 17**; **b → int 16**.

```python
a = 16  # Associa la variabile all’oggetto indicato.
b = a  # Associa un’altra etichetta allo stesso oggetto.

risultato = a + 1  # Valuta l’espressione e associa la variabile al risultato.
a = risultato  # Associa un’altra etichetta allo stesso oggetto.

print(a)  # Mostra il valore o il risultato indicato.
print(b)  # Mostra il valore o il risultato indicato.
```

## Modificare un oggetto condiviso

`valori` e `alias` fanno riferimento alla stessa lista. La modifica è visibile attraverso entrambe le etichette.

```python
valori = [10, 20]  # Associa la variabile all’oggetto indicato.
alias = valori  # Associa un’altra etichetta allo stesso oggetto.
alias.append(30)  # Aggiunge un elemento alla lista condivisa.

print(valori)  # Mostra il valore o il risultato indicato.
print(alias)  # Mostra il valore o il risultato indicato.
print(valori is alias)  # Confronta l’identità degli oggetti.
```

## Operatori aritmetici
Gli operatori lavorano sugli oggetti operando e producono un risultato.

| Operatore | Operazione |
|---|---|
| `+` | somma |
| `-` | differenza |
| `*` | prodotto |
| `/` | divisione |
| `//` | divisione con arrotondamento verso il basso |
| `%` | resto |
| `**` | potenza |

Con divisore zero, divisione e resto causano un errore.

```python
a = 7  # Associa la variabile all’oggetto indicato.
b = 2  # Associa la variabile all’oggetto indicato.

somma = a + b  # Valuta l’espressione e associa la variabile al risultato.
differenza = a - b  # Valuta l’espressione e associa la variabile al risultato.
prodotto = a * b  # Valuta l’espressione e associa la variabile al risultato.
divisione = a / b  # Valuta l’espressione e associa la variabile al risultato.
div_intera = a // b  # Valuta l’espressione e associa la variabile al risultato.
resto = a % b  # Valuta l’espressione e associa la variabile al risultato.
potenza = a ** b  # Valuta l’espressione e associa la variabile al risultato.

print(somma, differenza, prodotto)  # Mostra il valore o il risultato indicato.
print(divisione, div_intera, resto)  # Mostra il valore o il risultato indicato.
print(potenza)  # Mostra il valore o il risultato indicato.
```

## Lo stesso simbolo, operazioni diverse
Con due interi, `+` esegue una somma. Con due stringhe, esegue una **concatenazione**, cioè unisce i testi. L’operazione dipende dai tipi degli oggetti.

```python
a = 10  # Associa la variabile all’oggetto indicato.
b = 5  # Associa la variabile all’oggetto indicato.
somma = a + b  # Valuta l’espressione e associa la variabile al risultato.
s1 = "10"  # Associa la variabile all’oggetto indicato.
s2 = "5"  # Associa la variabile all’oggetto indicato.
unione = s1 + s2  # Valuta l’espressione e associa la variabile al risultato.
prima = "Py"  # Associa la variabile all’oggetto indicato.
seconda = "thon"  # Associa la variabile all’oggetto indicato.
parola = prima + seconda  # Valuta l’espressione e associa la variabile al risultato.
print(somma)  # Mostra il valore o il risultato indicato.
print(unione)  # Mostra il valore o il risultato indicato.
print(parola)  # Mostra il valore o il risultato indicato.
```

## Precedenza degli operatori
Le parentesi possono cambiare l’ordine di valutazione. Negli esempi: prima le potenze, poi prodotti e divisioni, infine somme e sottrazioni.

```python
a = 2  # Associa la variabile all’oggetto indicato.
b = 3  # Associa la variabile all’oggetto indicato.
c = 4  # Associa la variabile all’oggetto indicato.

risultato1 = a + b * c  # Valuta l’espressione e associa la variabile al risultato.
risultato2 = (a + b) * c  # Valuta l’espressione e associa la variabile al risultato.

base = 2  # Associa la variabile all’oggetto indicato.
esponente = 3  # Associa la variabile all’oggetto indicato.
fattore = 2  # Associa la variabile all’oggetto indicato.
risultato3 = base ** esponente * fattore  # Valuta l’espressione e associa la variabile al risultato.

print(risultato1)  # Mostra il valore o il risultato indicato.
print(risultato2)  # Mostra il valore o il risultato indicato.
print(risultato3)  # Mostra il valore o il risultato indicato.
```

## 4. Il rettangolo
**Consegna**

Acquisire base e altezza come `float`. Calcolare e visualizzare area e perimetro.

**Procedimento**

L’area è base per altezza; il perimetro è il doppio della loro somma. I risultati vengono associati a `area` e `perimetro`.

**Dati della sessione di esempio:** `5`, `3`.

```python
base = float(input("Base: "))  # Legge una stringa e la converte nel tipo richiesto.
altezza = float(input("Altezza: "))  # Legge una stringa e la converte nel tipo richiesto.
area = base * altezza  # Valuta l’espressione e associa la variabile al risultato.
perimetro = 2 * (base + altezza)  # Valuta l’espressione e associa la variabile al risultato.
print("Area:", area)  # Mostra il valore o il risultato indicato.
print("Perimetro:", perimetro)  # Mostra il valore o il risultato indicato.
```

All’inizio `a` indica l’intero 10. Dopo `b = a`, due etichette indicano lo stesso oggetto. L’espressione `a + incremento` produce il risultato 15. Dopo `a = risultato`, `a` indica 15 e `b` continua a indicare 10. L’output è quindi 15 e 10.

## 6. Individuare e correggere un errore
**Consegna**

Individuare i tipi di `x` e `y`, spiegare l’errore e correggere il codice per ottenere prima il numero 15 e poi la stringa `"105"`.

**Analisi**

`x` indica una stringa e `y` un intero. L’espressione `x + y` non è definita per questa combinazione di tipi. La cella seguente riproduce l’errore intenzionale.

```python
x = "10"  # Associa la variabile all’oggetto indicato.
y = 5  # Associa la variabile all’oggetto indicato.

risultato = x + y  # Valuta l’espressione e associa la variabile al risultato.

print(risultato)  # Mostra il valore o il risultato indicato.
```

### Soluzione numerica
La conversione della stringa in `int` permette la somma fra interi.

```python
x = "10"  # Associa la variabile all’oggetto indicato.
y = 5  # Associa la variabile all’oggetto indicato.
intero = int(x)  # Converte il dato e associa la variabile al risultato.
risultato = intero + y  # Valuta l’espressione e associa la variabile al risultato.
print(risultato)  # Mostra il valore o il risultato indicato.
```

### Soluzione testuale
La conversione dell’intero in `str` permette la concatenazione fra stringhe.

```python
x = "10"  # Associa la variabile all’oggetto indicato.
y = 5  # Associa la variabile all’oggetto indicato.
testo = str(y)  # Converte il dato e associa la variabile al risultato.
risultato = x + testo  # Valuta l’espressione e associa la variabile al risultato.
print(risultato)  # Mostra il valore o il risultato indicato.
```
