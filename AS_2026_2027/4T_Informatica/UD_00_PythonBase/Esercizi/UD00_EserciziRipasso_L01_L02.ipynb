## Come contare gli oggetti negli esercizi

Una **variabile** è un’etichetta che fa riferimento a un oggetto.

### La metrica didattica

Contiamo gli oggetti associati alle variabili durante l’esecuzione delle righe effettivamente eseguite:

- Un’assegnazione di un valore, come `a = 10`, introduce un oggetto nel modello.
- Un risultato assegnato a una variabile, come in `a = a + 5`, introduce l’oggetto risultato.
- Un’assegnazione come `b = a` aggiunge un’etichetta allo stesso oggetto, senza introdurne un altro.
- **Non contiamo i letterali usati soltanto come operandi**, come il `5` in `a + 5` o il `6` in `voto >= 6`.
- **Non contiamo i risultati temporanei non assegnati**, come i booleani usati soltanto nelle condizioni degli `if`.
- Un booleano assegnato, come in `sufficiente = voto >= 6`, viene invece contato.
- Non contiamo valori e risultati di righe o parti di espressioni non eseguite.

Questo è un modello didattico: non descrive le effettive allocazioni o ottimizzazioni di memoria di Python.

### Esempio

```python
a = 10
b = a
a = a + 5
```

Contiamo **2 oggetti**: l’intero `10`, indicato alla fine da `b`, e l’intero `15`, risultato della somma e indicato da `a`. Il `5` usato nella somma non compare nella tabella e non viene contato.

### Oggetti indicati dalle variabili alla fine

Contiamo gli **oggetti distinti con almeno un’etichetta alla fine**. Un oggetto con più etichette conta una sola volta. Un oggetto contato in precedenza ma non più indicato da variabili resta nel conteggio complessivo e viene segnalato come privo di riferimento diretto finale.

Per due assegnazioni indipendenti:

```python
a = 10
b = 10
```

Nel modello didattico contiamo **2 oggetti**, uno per ciascuna assegnazione indipendente, senza considerare l’eventuale riuso di oggetti da parte di Python.

## A1 · Riassegnazione e aliasing — svolto

````python
a = 10  # Creo l'oggetto intero 10 e l'etichetta/variabile a punterà a tale oggetto.
b = a  # Associa b allo stesso oggetto: nessun nuovo oggetto dati.
a = a + 5  # Viene creto un nuovo oggetto intero 15, e a punterà a questo nuovo oggetto. b rimane puntato all'oggetto 10.
print(a, b)  # Mostra 15 e 10
````

### Soluzione completa di A1

**Output finale**
```text
15 10
```

**Oggetti secondo la metrica didattica: 2.**

| Oggetto | Origine | Tipo | Valore | Etichette alla fine |
|---|---|---|---|---|
| O1 | intero `10` | `int` | 10 | `b` |
| O2 | risultato di `a + 5` | `int` | 15 | `a` |

**Oggetti distinti con riferimento diretto alla fine: 2.** Le etichette sono riportate nella tabella.

**Oggetti del modello senza riferimento diretto alla fine: nessuno.**

`b = a` aggiunge un’etichetta allo stesso oggetto. La riassegnazione di `a` lascia `b` associata a `10`. Il letterale `5` della somma è escluso.

## A2 · Stringhe, metodi e conversioni

````python
testo = "12"  # Creo la stringa "12"; testo punta a questo oggetto.
alias = testo  # Associo alias allo stesso oggetto: nessun nuovo oggetto dati.
numero = int(testo)  # Creo l'intero 12 tramite conversione; numero punta a questo oggetto.
risultato = numero + 3  # Creo l'intero 15; risultato punta a 15 e numero rimane su 12.
messaggio = "ciao"  # Creo la stringa "ciao"; messaggio punta a questo oggetto.
maiuscolo = messaggio.upper()  # Creo la stringa "CIAO"; la stringa originale non cambia.
print(alias, risultato)  # Mostra 12 e 15, separati da uno spazio.
print(messaggio, maiuscolo)  # Mostra ciao e CIAO, separati da uno spazio.
````

### Soluzione completa di A2

**Output finale**
```text
12 15
ciao CIAO
```

**Oggetti secondo la metrica didattica: 5.**

| Oggetto | Origine | Tipo | Valore | Etichette alla fine |
|---|---|---|---|---|
| O1 | stringa letterale | `str` | `"12"` | `testo`, `alias` |
| O2 | conversione int(testo) | `int` | `12` | `numero` |
| O3 | risultato numero + 3 | `int` | `15` | `risultato` |
| O4 | stringa letterale | `str` | `"ciao"` | `messaggio` |
| O5 | risultato messaggio.upper() | `str` | `"CIAO"` | `maiuscolo` |

**Oggetti distinti con riferimento diretto alla fine: 5.** Le etichette sono riportate nella tabella.

**Oggetti del modello senza riferimento diretto alla fine: nessuno.**

`testo` e `alias` indicano la stessa stringa. `.upper()` produce la stringa `"CIAO"` senza modificare `"ciao"`. Il letterale `3` della somma è escluso.

## A3 · Scelta alternativa

````python
voto = 5  # Creo l'intero 5; voto punta a questo oggetto.
copia = voto  # Associo copia allo stesso oggetto 5: nessun nuovo oggetto dati.
sufficiente = voto >= 6  # Il confronto produce False; sufficiente punta al risultato booleano.
if sufficiente:  # La condizione è falsa: non eseguo il ramo if.
    esito = "Superata"  # Riga non eseguita: non conto questa stringa nel modello.
else:  # Eseguo il ramo alternativo.
    esito = "Da recuperare"  # Creo la stringa "Da recuperare"; esito punta a questo oggetto.
voto = voto + 1  # Creo l'intero 6; voto cambia riferimento, copia rimane su 5.
print(esito)  # Mostra Da recuperare.
print(voto, copia)  # Mostra 6 e 5.
````

### Soluzione completa di A3

**Output finale**
```text
Da recuperare
6 5
```

**Oggetti secondo la metrica didattica: 4.**

| Oggetto | Origine | Tipo | Valore | Etichette alla fine |
|---|---|---|---|---|
| O1 | letterale iniziale | `int` | `5` | `copia` |
| O2 | risultato voto >= 6 | `bool` | `False` | `sufficiente` |
| O3 | stringa del ramo else | `str` | `"Da recuperare"` | `esito` |
| O4 | risultato voto + 1 | `int` | `6` | `voto` |

**Oggetti distinti con riferimento diretto alla fine: 4.** Le etichette sono riportate nella tabella.

**Oggetti del modello senza riferimento diretto alla fine: nessuno.**

`False` viene contato perché è assegnato a `sufficiente`. I letterali `6` del confronto e `1` della somma sono esclusi; l’intero `6` risultato della somma viene contato perché è assegnato a `voto`. Il ramo che assegna `"Superata"` non viene eseguito.

## A4 · Selezione multipla

````python
punteggio = 8  # Creo l'intero 8; punteggio punta a questo oggetto.
if punteggio >= 9:  # Il confronto produce False: salto questo ramo.
    categoria = "Alto"  # Riga non eseguita.
elif punteggio >= 6:  # Il confronto produce True: eseguo questo ramo.
    categoria = "Base"  # Creo la stringa "Base"; categoria punta a questo oggetto.
else:  # Ramo saltato, perché la condizione elif era vera.
    categoria = "Basso"  # Riga non eseguita.
alias = categoria  # Associo alias allo stesso oggetto "Base".
categoria = categoria.upper()  # Creo "BASE"; categoria cambia riferimento, alias rimane su "Base".
print(alias)  # Mostra Base.
print(categoria)  # Mostra BASE.
````

### Soluzione completa di A4

**Output finale**
```text
Base
BASE
```

**Oggetti secondo la metrica didattica: 3.**

| Oggetto | Origine | Tipo | Valore | Etichette alla fine |
|---|---|---|---|---|
| O1 | letterale iniziale | `int` | `8` | `punteggio` |
| O2 | stringa del ramo elif | `str` | `"Base"` | `alias` |
| O3 | risultato categoria.upper() | `str` | `"BASE"` | `categoria` |

**Oggetti distinti con riferimento diretto alla fine: 3.** Le etichette sono riportate nella tabella.

**Oggetti del modello senza riferimento diretto alla fine: nessuno.**

I letterali `9` e `6` dei confronti e i risultati booleani usati soltanto da `if` ed `elif` sono esclusi. I rami che assegnano `"Alto"` e `"Basso"` non vengono eseguiti. `alias` conserva `"Base"` mentre `categoria` indica il nuovo risultato `"BASE"`.

## A5 · Condizioni indipendenti e corto circuito

````python
divisore = 0  # Creo l'intero 0; divisore punta a questo oggetto.
riferimento = divisore  # Associo riferimento allo stesso oggetto 0.
if divisore != 0 and 12 / divisore > 2:  # Il primo confronto è False: and non valuta la divisione.
    esito = "Quoziente alto"  # Riga non eseguita.
else:  # Eseguo il ramo alternativo.
    esito = "Calcolo escluso"  # Creo la stringa "Calcolo escluso"; esito punta a questo oggetto.
if divisore == 0:  # Il confronto produce True: questo if è indipendente dal precedente.
    divisore = 3  # Creo l'intero 3; divisore cambia riferimento, riferimento rimane su 0.
print(esito)  # Mostra Calcolo escluso.
print(divisore, riferimento)  # Mostra 3 e 0.
````

### Soluzione completa di A5

**Output finale**
```text
Calcolo escluso
3 0
```

**Oggetti secondo la metrica didattica: 3.**

| Oggetto | Origine | Tipo | Valore | Etichette alla fine |
|---|---|---|---|---|
| O1 | letterale iniziale | `int` | `0` | `riferimento` |
| O2 | stringa del ramo else | `str` | `"Calcolo escluso"` | `esito` |
| O3 | letterale nel secondo if | `int` | `3` | `divisore` |

**Oggetti distinti con riferimento diretto alla fine: 3.** Le etichette sono riportate nella tabella.

**Oggetti del modello senza riferimento diretto alla fine: nessuno.**

Gli zeri usati soltanto nei confronti e i risultati booleani delle condizioni sono esclusi. Il corto circuito di `and` evita la divisione per zero e la valutazione della seconda condizione. `riferimento` conserva lo zero iniziale; il `3` viene contato perché assegnato direttamente a `divisore`. Il ramo che assegna `"Quoziente alto"` non viene eseguito.

## B1 · Sconto su un acquisto

**Problema.** Acquisire un importo, ipotizzando sia non negativo. Applicare uno sconto del 10% se l’importo è almeno 100 euro; negli altri casi lo sconto è zero. Calcolare e mostrare lo sconto e il totale da pagare. Si assume un input valido.

**Esempio matematico:** importo 120 euro → sconto = 120 × 10 / 100 = 12 euro; totale = 120 − 12 = 108 euro.

````python
importo = float(input("Importo: "))  # Legge il testo e lo converte in un numero decimale.
if importo >= 100:  # Verifica se è stata raggiunta la soglia per lo sconto.
    sconto = importo * 10 / 100  # Calcola il 10% dell’importo.
else:  # Gestisce gli importi inferiori alla soglia.
    sconto = 0.0  # Assegna uno sconto nullo.
totale = importo - sconto  # Calcola quanto rimane da pagare.
print("Sconto:", sconto)  # Mostra lo sconto calcolato.
print("Totale:", totale)  # Mostra l’importo finale.
````

## B2 · Conversione di temperatura

**Problema.** Acquisire una temperatura in gradi Celsius, ipotizzando sia un valore numerico valido, e convertirla in Fahrenheit con la formula F = C × 9 / 5 + 32. Mostrare il risultato e il messaggio “Sotto zero” se C è inferiore a 0, altrimenti “Zero o superiore”.

**Esempio matematico:** C = −5 → F = −5 × 9 / 5 + 32 = 23. Risultato: 23 °F e “Sotto zero”.

````python
celsius = float(input("Temperatura in gradi Celsius: "))  # Legge il testo e lo converte in un numero decimale.
fahrenheit = celsius * 9 / 5 + 32  # Applica la formula di conversione.
print("Temperatura in Fahrenheit:", fahrenheit)  # Mostra la temperatura convertita.
if celsius < 0:  # Verifica se la temperatura iniziale è negativa.
    print("Sotto zero")  # Mostra il messaggio per le temperature negative.
else:  # Gestisce lo zero e le temperature positive.
    print("Zero o superiore")  # Mostra il messaggio richiesto.
````

## B3 · Media tra 5 numeri interi

**Problema.** Acquisire cinque numeri interi. Calcolare la media aritmetica. Mostrare la media e una sola fascia: “Alta” se la media è almeno 8; “Sufficiente” da 6 incluso a 8 escluso; “Insufficiente” sotto 6. Si assumono numeri validi.

**Esempio matematico:** Voti 7, 9, 8, 10, 6 → media = (7 + 9 + 8 + 10 + 6) / 5 = 8. Risultato: media 8 e fascia “Alta”.

````python
numero1 = int(input("Primo numero: "))  # Legge e converte il primo numero intero.
numero2 = int(input("Secondo numero: "))  # Legge e converte il secondo numero intero.
numero3 = int(input("Terzo numero: "))  # Legge e converte il terzo numero intero.
numero4 = int(input("Quarto numero: "))  # Legge e converte il quarto numero intero.
numero5 = int(input("Quinto numero: "))  # Legge e converte il quinto numero intero.
media = (numero1 + numero2 + numero3 + numero4 + numero5) / 5  # Calcola la media aritmetica.
print("Media:", media)  # Mostra la media calcolata.
if media >= 8:  # Controlla per prima la fascia più alta.
    print("Alta")  # Mostra la fascia per medie almeno pari a 8.
elif media >= 6:  # Qui la media è già inferiore a 8: basta controllare la soglia 6.
    print("Sufficiente")  # Mostra la fascia da 6 incluso a 8 escluso.
else:  # Gestisce le medie inferiori a 6.
    print("Insufficiente")  # Mostra la fascia più bassa.
````

## B4 · Divisione controllata

**Problema.** Acquisire due numeri, ipotizzando siano validi, il dividendo e il divisore. Se il divisore è diverso da zero, calcolare e mostrare il quoziente; se è zero, mostrare “Divisione impossibile” senza eseguire la divisione.

**Esempio matematico:** 12 e 4 → 12 / 4 = 3. Con 12 e 0 il risultato richiesto è il messaggio “Divisione impossibile”.

````python
dividendo = float(input("Dividendo: "))  # Legge e converte il dividendo.
divisore = float(input("Divisore: "))  # Legge e converte il divisore.
if divisore != 0:  # Controlla il divisore prima di eseguire la divisione.
    quoziente = dividendo / divisore  # Calcola il quoziente solo con divisore diverso da zero.
    print("Quoziente:", quoziente)  # Mostra il risultato della divisione.
else:  # Gestisce il divisore uguale a zero.
    print("Divisione impossibile")  # Mostra il messaggio senza eseguire la divisione.
````

## B5 · Calcolo area e verifica quadrato

**Problema.** Acquisire base e altezza di un rettangolo. Se almeno una misura è minore o uguale a zero, mostrare “Misure non valide”. Altrimenti calcolare area e perimetro e mostrare “Quadrato” quando le due misure sono uguali, oppure “Rettangolo non quadrato” quando sono diverse. Si assumono input numerici convertibili.

````python
base = float(input("Base: "))  # Legge e converte la base.
altezza = float(input("Altezza: "))  # Legge e converte l'altezza.
if base <= 0 or altezza <= 0:  # Basta una misura non positiva per rendere i dati non validi.
    print("Misure non valide")  # Mostra il messaggio senza calcolare area e perimetro.
else:  # Esegue i calcoli solo quando entrambe le misure sono positive.
    area = base * altezza  # Calcola l'area del rettangolo.
    perimetro = (base + altezza) * 2  # Calcola la somma delle lunghezze dei quattro lati.
    print("Area:", area)  # Mostra l'area.
    print("Perimetro:", perimetro)  # Mostra il perimetro.
    if base == altezza:  # Verifica se le due misure sono uguali.
        print("Quadrato")  # Mostra il caso particolare del quadrato.
    else:  # Gestisce le misure positive ma diverse.
        print("Rettangolo non quadrato")  # Mostra il tipo di figura.
````
