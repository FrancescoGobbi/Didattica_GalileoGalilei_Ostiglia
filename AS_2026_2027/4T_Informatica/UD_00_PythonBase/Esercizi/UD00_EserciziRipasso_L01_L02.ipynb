## Come contare gli oggetti negli esercizi

Una **variabile** è un’etichetta che fa riferimento a un oggetto.

Con `b = a`, le due variabili indicano **lo stesso oggetto**: non viene creato un altro oggetto.

### Il conteggio usato dal punto di vista didattico

Python, per ottimizzare la memoria, può riutilizzare alcuni oggetti. Per rendere gli esercizi semplici, usiamo quindi un **modello didattico**: contiamo gli oggetti che le espressioni potrebbero produrre, non le vere creazioni in memoria.

### Esempio

```python
a = 10
b = a
a = a + 5
```

Nel nostro modello contiamo **2 oggetti potenziali**:

- l’intero `10`;
- l’intero `15`, risultato della somma.
- (per la correttezza dell’esempio, si può anche contare l’intero `5`, ma non è necessario).

L’istruzione `b = a` non aggiunge un oggetto.

### Quali oggetti sono indicati dalle variabili alla fine?

Contiamo gli **oggetti distinti indicati direttamente dalle variabili** e scriviamo quali etichette li indicano. Se due variabili indicano lo stesso oggetto, viene contato una sola volta.

Nell’esempio:

- `a` indica l’intero `15`;
- `b` indica l’intero `10`;
- nessuna variabile dell’esercizio indica direttamente l’intero `5`.

Quindi, alla fine, **2 oggetti hanno un’etichetta diretta**.

Ulteriore esempio:

```python
a = 10
b = 10
```

Nel nostro modello contiamo **2 oggetti potenziali**: l’intero `10` crato due volte. 
All'effettivo, quindi per l'interprete di Python che ottimizza la memoria, **c'è un solo oggetto `10` in memoria**, ma il nostro modello didattico conta due oggetti distinti.

## A1 · Riassegnazione e aliasing — svolto

````python
a = 10  # Creo l'oggetto intero 10 e l'etichetta/variabile a punterà a tale oggetto.
b = a  # Associa b allo stesso oggetto: nessun nuovo oggetto dati.
a = a + 5  # Viene creto un nuovo oggetto intero 15, e a punterà a questo nuovo oggetto. b rimane puntato all'oggetto 10.
print(a, b)  # Mostra 15 e 10
````

### Soluzione completa di A1

**Output finale**
```text
15 10
```

**Oggetti dati potenziali: 2.**

| Oggetto | Origine | Tipo | Valore | Etichette alla fine |
|---|---|---|---|---|
| O1 | intero `10` | `int` | 10 | `b` |
| O3 | risultato di `a + 5` | `int` | 15 | `a` |

## B1 · Sconto su un acquisto

**Problema.** Acquisire un importo, ipotizzando sia non negativo. Applicare uno sconto del 10% se l’importo è almeno 100 euro; negli altri casi lo sconto è zero. Calcolare e mostrare lo sconto e il totale da pagare. Si assume un input valido.

**Esempio matematico:** importo 120 euro → sconto = 120 × 10 / 100 = 12 euro; totale = 120 − 12 = 108 euro.

````python
importo = float(input("Importo: "))  # Legge il testo e lo converte in un numero decimale.
if importo >= 100:  # Verifica se è stata raggiunta la soglia per lo sconto.
    sconto = importo * 10 / 100  # Calcola il 10% dell’importo.
else:  # Gestisce gli importi inferiori alla soglia.
    sconto = 0.0  # Assegna uno sconto nullo.
totale = importo - sconto  # Calcola quanto rimane da pagare.
print("Sconto:", sconto)  # Mostra lo sconto calcolato.
print("Totale:", totale)  # Mostra l’importo finale.
````
