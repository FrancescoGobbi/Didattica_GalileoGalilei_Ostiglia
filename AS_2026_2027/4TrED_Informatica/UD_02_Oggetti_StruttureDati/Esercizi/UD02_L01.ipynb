## Le liste in sintesi

| Concetto | Significato | Esempio |
|---|---|---|
| Lista | Oggetto della classe `list` che contiene una **sequenza ordinata** di riferimenti ad altri oggetti | `voti = [7, 8, 6]` |
| Elemento | Ogni oggetto indicato da una cella della lista | `7`, `8` e `6` |
| Indice | Posizione di un elemento: parte da **0** a sinistra e da **−1** a destra | `voti[0]` vale 7, `voti[-1]` vale 6 |
| Lunghezza | Numero di elementi, restituito da `len()` | `len(voti)` vale 3 |
| Mutabile | La lista può essere **modificata** dopo la creazione: resta lo stesso oggetto, con lo stesso ID | `voti[1] = 9` |
| Metodo | Funzione che appartiene a una classe e si chiama sull'oggetto con il punto | `voti.append(5)` |

**Nota bene.** La lista non contiene gli oggetti, ma i **riferimenti** agli oggetti: ogni cella indica un oggetto che si trova altrove in memoria, proprio come fa un'etichetta.

## Lo schema in memoria

Dopo l'istruzione `voti = [7, 8, 6]` la situazione in memoria è la seguente.

```text
 etichetta            oggetto list                  oggetti int
                    ┌─────┬─────┬─────┐
   voti  ────────►  │  ●  │  ●  │  ●  │
                    └──┼──┴──┼──┴──┼──┘
   indici:            0     1     2
                       │     │     │
                       ▼     ▼     ▼
                       7     8     6
```

L'etichetta `voti` indica l'oggetto lista; ogni cella della lista indica un oggetto `int`.

## 1. Creare una lista e riconoscerne la classe

Una lista si crea scrivendo gli elementi tra **parentesi quadre**, separati da virgole. La funzione `type()` conferma la classe `list`, mentre `len()` restituisce il numero di elementi.

**Output atteso:** `[7, 8, 6]`, `<class 'list'>` e `3`.

### Versione semplificata

````python
def main():  # Definisce il programma principale.
    voti = [7, 8, 6]  # Viene creato un oggetto list con tre riferimenti.
    print(voti)  # Viene mostrata la lista.
    print(type(voti))  # Viene mostrata la classe dell'oggetto.
    print(len(voti))  # Viene mostrato il numero di elementi.

if __name__ == '__main__':  # Controlla l'avvio diretto.
    main()  # Avvia il programma.
````

### Versione con ID

````python
def main():  # Definisce il programma principale.
    voti = [7, 8, 6]  # Viene creato un oggetto list con tre riferimenti.
    print("Lista:", voti, "· ID della lista:", id(voti))  # La lista è un oggetto con il proprio ID.
    for i in range(len(voti)):  # Vengono percorsi gli indici 0, 1, 2.
        print("Cella", i, "→ oggetto", voti[i], "· ID:", id(voti[i]))  # Ogni cella indica un oggetto distinto.

if __name__ == '__main__':  # Controlla l'avvio diretto.
    main()  # Avvia il programma.
````

**Osservazione.** Nella versione con ID compaiono **quattro** ID diversi: uno per l'oggetto lista e uno per ciascun oggetto `int`. La lista e gli interi sono oggetti separati, collegati dai riferimenti.

## 2. Quali oggetti può contenere una lista

Una lista può contenere oggetti di **qualsiasi classe**: `int`, `float`, `str`, `bool`, `None` e perfino **altre liste**. Gli elementi possono essere **di tipo diverso** nella stessa lista: la lista non ha un tipo per gli elementi, ogni cella indica un oggetto con la **propria classe**.

**Output atteso:** sei righe, una per elemento, con il valore e la sua classe.

````python
def main():  # Definisce il programma principale.
    dati = [16, 7.5, "Anna", True, None, [1, 2]]  # Sei oggetti di classi diverse.
    for elemento in dati:  # Viene percorsa la lista per valore.
        print(elemento, type(elemento))  # Viene mostrato il valore con la sua classe.

if __name__ == '__main__':  # Controlla l'avvio diretto.
    main()  # Avvia il programma.
````

## 2.1 Una lista dentro una lista

Quando un elemento è a sua volta una lista, si usano **due indici**: il primo sceglie l'elemento della lista esterna, il secondo l'elemento della lista interna.

**Output atteso:** `['Anna', [7, 8, 6]]`, `[7, 8, 6]`, `8` e `2`.

````python
def main():  # Definisce il programma principale.
    studente = ["Anna", [7, 8, 6]]  # Il secondo elemento è una lista di voti.
    print(studente)  # Viene mostrata la lista esterna.
    print(studente[1])  # Viene mostrata la lista interna.
    print(studente[1][1])  # Secondo voto della lista interna.
    print(len(studente))  # La lista esterna ha due elementi.

if __name__ == '__main__':  # Controlla l'avvio diretto.
    main()  # Avvia il programma.
````

**Nota bene.** `len(studente)` vale 2 e non 4: la lista interna conta come **un solo elemento** della lista esterna. Di solito conviene usare elementi **dello stesso tipo**, perché funzioni come `sum()`, `max()` e `min()` li richiedono.

## 3. I modi per creare una lista

Oltre alle parentesi quadre, la funzione `list()` costruisce una lista a partire da un'altra sequenza, per esempio un `range()` o una stringa.

| Istruzione | Lista ottenuta |
|---|---|
| `[]` | lista vuota |
| `[7, 8, 6]` | tre interi |
| `["Anna", 16, True]` | tipi diversi |
| `list(range(1, 6))` | `[1, 2, 3, 4, 5]` |
| `list(range(10, 0, -3))` | `[10, 7, 4, 1]` |
| `list("ciao")` | `['c', 'i', 'a', 'o']` |

````python
def main():  # Definisce il programma principale.
    vuota = []  # Lista senza elementi.
    voti = [7, 8, 6]  # Tre interi.
    misti = ["Anna", 16, True]  # Tipi diversi.
    numeri = list(range(1, 6))  # Lista ottenuta da un intervallo.
    alla_rovescia = list(range(10, 0, -3))  # Intervallo con passo negativo.
    lettere = list("ciao")  # Ogni carattere diventa un elemento.
    print(vuota, len(vuota))  # La lista vuota ha lunghezza 0.
    print(voti, misti)  # Vengono mostrate le liste scritte a mano.
    print(numeri, alla_rovescia)  # Vengono mostrate le liste ottenute dai range.
    print(lettere)  # Viene mostrata la lista di caratteri.

if __name__ == '__main__':  # Controlla l'avvio diretto.
    main()  # Avvia il programma.
````

## 4. Leggere un elemento con l'indice

`lista[i]` restituisce l'elemento in posizione `i`. Gli **indici positivi** partono da 0 a sinistra, gli **indici negativi** partono da −1 a destra.

```text
 elemento          7    8    6    9    5
 indice positivo   0    1    2    3    4
 indice negativo  -5   -4   -3   -2   -1
```

L'ultimo indice positivo valido è **`len(voti) - 1`**.

**Output atteso:** `7 6`, `5 9`, `5` e `5`.

### Versione semplificata

````python
def main():  # Definisce il programma principale.
    voti = [7, 8, 6, 9, 5]  # Lista di cinque voti.
    print(voti[0], voti[2])  # Primo e terzo elemento.
    print(voti[-1], voti[-2])  # Ultimo e penultimo elemento.
    print(len(voti))  # Numero di elementi.
    print(voti[len(voti) - 1])  # Ultimo elemento letto con l'indice positivo.

if __name__ == '__main__':  # Controlla l'avvio diretto.
    main()  # Avvia il programma.
````

### Versione con ID

````python
def main():  # Definisce il programma principale.
    voti = [7, 8, 6, 9, 5]  # Lista di cinque voti.
    print("voti[0]  =", voti[0], "· ID:", id(voti[0]))  # Primo elemento letto da sinistra.
    print("voti[-5] =", voti[-5], "· ID:", id(voti[-5]))  # Lo stesso elemento letto da destra.
    print("voti[4]  =", voti[4], "· ID:", id(voti[4]))  # Ultimo elemento letto da sinistra.
    print("voti[-1] =", voti[-1], "· ID:", id(voti[-1]))  # Lo stesso elemento letto da destra.

if __name__ == '__main__':  # Controlla l'avvio diretto.
    main()  # Avvia il programma.
````

**Osservazione.** `voti[0]` e `voti[-5]` mostrano lo **stesso ID**: i due indici indicano la stessa cella e quindi lo stesso oggetto. Vale lo stesso per `voti[4]` e `voti[-1]`.

## 4.1 La sezione di una lista

`lista[inizio:fine]` restituisce una **nuova lista** con gli elementi dall'indice `inizio` all'indice `fine - 1`: come in `range()`, il valore finale è **escluso**. Se `inizio` manca si parte da 0; se `fine` manca si arriva in fondo.

**Output atteso:** `[8, 6]`, `[7, 8]`, `[9, 5]` e `False`.

````python
def main():  # Definisce il programma principale.
    voti = [7, 8, 6, 9, 5]  # Lista di cinque voti.
    centrali = voti[1:3]  # Indici 1 e 2: il 3 è escluso.
    primi = voti[:2]  # Dall'inizio fino all'indice 1.
    ultimi = voti[3:]  # Dall'indice 3 fino in fondo.
    print(centrali)  # Viene mostrata la prima sezione.
    print(primi)  # Viene mostrata la seconda sezione.
    print(ultimi)  # Viene mostrata la terza sezione.
    print(id(centrali) == id(voti))  # La sezione è un oggetto lista diverso.

if __name__ == '__main__':  # Controlla l'avvio diretto.
    main()  # Avvia il programma.
````

## 4.2 Indice fuori dall'intervallo

Con cinque elementi gli indici validi vanno da 0 a 4. L'indice 5 non corrisponde ad alcuna cella.

**Nota bene.** La cella produce volutamente un errore: compare `7`, poi un messaggio che termina con `IndexError: list index out of range`.

````python
def main():  # Definisce il programma principale.
    voti = [7, 8, 6, 9, 5]  # Indici validi: da 0 a 4.
    print(voti[0])  # Primo elemento: indice valido.
    print(voti[5])  # Errore: l'indice 5 non esiste.

if __name__ == '__main__':  # Controlla l'avvio diretto.
    main()  # Avvia il programma.
````

## 5. Cambiare un elemento della lista

L'assegnazione `voti[1] = 9` sostituisce il **riferimento** contenuto nella cella 1. L'intero 8 non viene modificato, perché gli `int` sono **immutabili**: la cella indica semplicemente un altro oggetto. La **lista**, invece, è stata modificata: è **mutabile** e resta lo stesso oggetto.

```text
 prima:   voti ──► [ ● , ● , ● ]        dopo:   voti ──► [ ● , ● , ● ]
                     │   │   │                             │   │   │
                     7   8   6                             7   9   6
```

**Output atteso:** `[7, 8, 6]` e poi `[7, 9, 6]`.

### Versione semplificata

````python
def main():  # Definisce il programma principale.
    voti = [7, 8, 6]  # Lista di tre voti.
    print(voti)  # Lista prima della modifica.
    voti[1] = 9  # La cella 1 indica un nuovo oggetto.
    print(voti)  # Lista dopo la modifica.

if __name__ == '__main__':  # Controlla l'avvio diretto.
    main()  # Avvia il programma.
````

### Versione con ID

````python
def main():  # Definisce il programma principale.
    voti = [7, 8, 6]  # Lista di tre voti.
    print("Prima:", voti, "· ID lista:", id(voti), "· ID cella 1:", id(voti[1]))  # Situazione iniziale.
    voti[1] = 9  # Cambia il riferimento nella cella 1.
    print("Dopo: ", voti, "· ID lista:", id(voti), "· ID cella 1:", id(voti[1]))  # Situazione finale.

if __name__ == '__main__':  # Controlla l'avvio diretto.
    main()  # Avvia il programma.
````

**Osservazione.** L'**ID della lista** è identico prima e dopo la modifica; l'**ID della cella 1** cambia, perché la cella indica ora l'oggetto 9 al posto dell'oggetto 8.

## 5.1 Modificare usando il valore precedente

Il nuovo valore di una cella può essere calcolato a partire da quello vecchio. Nell'esempio il primo voto aumenta di 1 e l'ultimo viene sostituito dal doppio del secondo.

**Output atteso:** `[8, 8, 16]`.

````python
def main():  # Definisce il programma principale.
    voti = [7, 8, 6]  # Lista di tre voti.
    voti[0] = voti[0] + 1  # Si legge 7, si calcola 8, la cella 0 indica 8.
    voti[-1] = voti[1] * 2  # L'ultima cella indica il doppio del secondo elemento.
    print(voti)  # Viene mostrata la lista modificata.

if __name__ == '__main__':  # Controlla l'avvio diretto.
    main()  # Avvia il programma.
````

**Nota bene.** Nell'istruzione `voti[0] = voti[0] + 1` Python calcola prima il lato destro, leggendo il valore attuale della cella, e solo dopo aggiorna il riferimento della cella.

## 6. Le funzioni `len()`, `sum()`, `max()` e `min()`

Alcune funzioni già note ricevono una lista come argomento e restituiscono un valore, **senza modificare** la lista.

| Funzione | Restituisce | Con `[7, 8, 6, 9]` |
|---|---|---|
| `len(lista)` | numero di elementi | 4 |
| `sum(lista)` | somma degli elementi numerici | 30 |
| `max(lista)` | elemento più grande | 9 |
| `min(lista)` | elemento più piccolo | 6 |

La **media** si ottiene dividendo la somma per il numero di elementi: 30 / 4 = 7.5.

````python
def main():  # Definisce il programma principale.
    voti = [7, 8, 6, 9]  # Lista di quattro voti.
    print(len(voti))  # Numero di elementi.
    print(sum(voti))  # Somma dei voti.
    print(max(voti), min(voti))  # Voto più alto e più basso.
    media = sum(voti) / len(voti)  # Somma divisa per il numero di elementi.
    print(media)  # Viene mostrata la media.
    print(voti)  # La lista non è stata modificata.

if __name__ == '__main__':  # Controlla l'avvio diretto.
    main()  # Avvia il programma.
````

## 6.1 Le stesse funzioni con altri tipi

`max()` e `min()` funzionano anche con le stringhe, confrontate in ordine alfabetico; `len()` funziona con qualsiasi lista. `sum()`, invece, richiede elementi **numerici**.

**Output atteso:** `3`, `Marco` e `Anna`, poi `10.5`.

````python
def main():  # Definisce il programma principale.
    nomi = ["Luca", "Anna", "Marco"]  # Lista di stringhe.
    print(len(nomi))  # Numero di nomi.
    print(max(nomi), min(nomi))  # Ultimo e primo in ordine alfabetico.
    prezzi = [2.5, 3.0, 5.0]  # Lista di numeri decimali.
    print(sum(prezzi))  # Somma dei prezzi.

if __name__ == '__main__':  # Controlla l'avvio diretto.
    main()  # Avvia il programma.
````

## 6.2 La media di una lista che potrebbe essere vuota

Con una lista **vuota** `len()` restituisce 0 e la divisione `sum(voti) / len(voti)` provocherebbe `ZeroDivisionError`. Prima di calcolare la media conviene quindi controllare che la lista contenga elementi.

**Output atteso:** `Media: 7.0` per la prima lista e `Nessun voto inserito` per la seconda.

````python
def mostra_media(voti):  # Riceve una lista di voti, anche vuota.
    if len(voti) > 0:  # La lista contiene almeno un elemento.
        print("Media:", sum(voti) / len(voti))  # La divisione è sicura.
    else:  # La lista è vuota.
        print("Nessun voto inserito")  # Viene evitata la divisione per zero.

def main():  # Definisce il programma principale.
    mostra_media([6, 8, 7])  # Lista con tre voti.
    mostra_media([])  # Lista vuota.

if __name__ == '__main__':  # Controlla l'avvio diretto.
    main()  # Avvia il programma.
````

## 7. Percorrere una lista con il ciclo `for`

Il ciclo `for` può percorrere una lista in due modi:

- **per valore**, `for voto in voti`, quando interessa solo l'elemento;
- **per indice**, `for i in range(len(voti))`, quando serve anche la posizione.

`range(len(voti))` fornisce esattamente gli indici validi: con tre elementi, 0, 1 e 2.

````python
def main():  # Definisce il programma principale.
    voti = [7, 8, 6]  # Lista di tre voti.
    print("Per valore:")  # Prima modalità.
    for voto in voti:  # voto viene associato a ogni elemento, in ordine.
        print(voto)  # Viene mostrato l'elemento.
    print("Per indice:")  # Seconda modalità.
    for i in range(len(voti)):  # i assume i valori 0, 1, 2.
        print("Posizione", i, "voto", voti[i])  # Vengono mostrati indice ed elemento.

if __name__ == '__main__':  # Controlla l'avvio diretto.
    main()  # Avvia il programma.
````

## 7.1 Contare gli elementi che rispettano una condizione

Un **contatore** inizializzato a 0 prima del ciclo viene aumentato di 1 ogni volta che un elemento rispetta la condizione. Nell'esempio vengono contati i voti **insufficienti**, cioè minori di 6.

**Output atteso:** `Insufficienze: 2`.

````python
def main():  # Definisce il programma principale.
    voti = [7, 5, 8, 4, 6]  # Lista di cinque voti.
    insufficienze = 0  # Il contatore parte da 0, prima del ciclo.
    for voto in voti:  # Viene percorsa la lista per valore.
        if voto < 6:  # Il voto è insufficiente.
            insufficienze = insufficienze + 1  # Il contatore aumenta di 1.
    print("Insufficienze:", insufficienze)  # Viene mostrato il conteggio.

if __name__ == '__main__':  # Controlla l'avvio diretto.
    main()  # Avvia il programma.
````

| Voto | `voto < 6` | Contatore |
|---|---|---|
| 7 | `False` | 0 |
| 5 | `True` | 1 |
| 8 | `False` | 1 |
| 4 | `True` | 2 |
| 6 | `False` | 2 |

## 7.2 Trovare la posizione del valore più alto

`max()` restituisce il valore più alto, ma non la sua posizione. Per conoscere anche l'indice serve un ciclo **per indice**, che ricorda la posizione del massimo trovato fino a quel momento.

**Output atteso:** `Massimo: 9 in posizione 3`.

````python
def main():  # Definisce il programma principale.
    voti = [7, 8, 6, 9, 5]  # Lista di cinque voti.
    posizione = 0  # Si suppone che il massimo sia il primo elemento.
    for i in range(1, len(voti)):  # Vengono controllati gli altri indici.
        if voti[i] > voti[posizione]:  # Trovato un elemento più grande.
            posizione = i  # Viene ricordata la nuova posizione.
    print("Massimo:", voti[posizione], "in posizione", posizione)  # Viene mostrato il risultato.

if __name__ == '__main__':  # Controlla l'avvio diretto.
    main()  # Avvia il programma.
````

**Osservazione.** Il ciclo parte dall'indice 1, perché l'elemento in posizione 0 è già il candidato iniziale. Il risultato si può verificare con `max(voti)`, che restituisce 9.

## 8. Il ciclo `for` e gli oggetti della lista

Nel ciclo `for voto in voti`, a ogni giro l'etichetta `voto` viene associata **allo stesso oggetto** indicato dalla cella: **non viene creata una copia**. Riassegnare `voto` cambia solo l'etichetta, **non la lista**. Per modificare un elemento serve l'**indice**: `voti[i] = ...`.

**Output atteso:** `[7, 8, 6]`, perché il primo ciclo non cambia la lista, e poi `[8, 9, 7]`.

### Versione semplificata

````python
def main():  # Definisce il programma principale.
    voti = [7, 8, 6]  # Lista di tre voti.
    for voto in voti:  # voto indica, a turno, gli oggetti della lista.
        voto = voto + 1  # Cambia solo l'etichetta voto.
    print(voti)  # La lista è invariata.
    for i in range(len(voti)):  # Vengono percorsi gli indici.
        voti[i] = voti[i] + 1  # Cambia il riferimento nella cella i.
    print(voti)  # La lista è modificata.

if __name__ == '__main__':  # Controlla l'avvio diretto.
    main()  # Avvia il programma.
````

### Versione con ID

````python
def main():  # Definisce il programma principale.
    voti = [7, 8, 6]  # Lista di tre voti.
    for voto in voti:  # voto indica, a turno, gli oggetti della lista.
        print("voto =", voto, "· ID:", id(voto))  # Stesso ID dell'oggetto nella cella.
        voto = voto + 1  # voto viene associato a un nuovo oggetto.
        print("   dopo voto + 1: voto =", voto, "· ID:", id(voto))  # L'ID è cambiato.
    print("Lista:", voti)  # La lista è invariata.
    for i in range(len(voti)):  # Vengono mostrati gli ID delle celle.
        print(f"voti[{i}] =", voti[i], "· ID:", id(voti[i]))  # Gli ID iniziali sono rimasti nelle celle.

if __name__ == '__main__':  # Controlla l'avvio diretto.
    main()  # Avvia il programma.
````

**Osservazione.** Il primo ID di ogni giro coincide con quello della cella: `voto` e la cella indicano lo stesso oggetto. Dopo `voto = voto + 1` l'ID cambia solo per `voto`; gli ID mostrati alla fine, cella per cella, sono quelli iniziali.

## 9. Che cos'è un metodo

Un **metodo** è una funzione che appartiene a una classe. Si chiama **sull'oggetto**, con la **notazione puntata**:

```text
   voti   .   append   (9)
    │           │       │
 oggetto     metodo  argomento
```

Un metodo è già stato usato nella lezione 01 dell'unità 01: `upper()` delle stringhe. Esiste però una differenza importante: le stringhe sono **immutabili**, quindi `upper()` restituisce una **nuova stringa**; le liste sono **mutabili**, quindi `append()` **modifica la lista stessa**.

| | Funzione | Metodo |
|---|---|---|
| Esempio | `len(voti)` | `voti.append(9)` |
| Dove sta la lista | tra le parentesi | prima del punto |
| Effetto sulla lista | `len()` non la modifica | `append()` e `insert()` la modificano |

````python
def main():  # Definisce il programma principale.
    nome = "anna"  # Oggetto str, immutabile.
    maiuscolo = nome.upper()  # Il metodo restituisce una nuova stringa.
    print(nome, maiuscolo)  # La stringa originale è invariata.
    voti = [7, 8]  # Oggetto list, mutabile.
    print(len(voti))  # Funzione: la lista è tra le parentesi.
    voti.append(9)  # Metodo: la lista è prima del punto e viene modificata.
    print(voti)  # La lista contiene ora tre elementi.

if __name__ == '__main__':  # Controlla l'avvio diretto.
    main()  # Avvia il programma.
````

## 10. Il metodo `append()`

`append(x)` aggiunge l'oggetto `x` **in fondo** alla lista. La lista cresce di un elemento e resta **lo stesso oggetto**.

| Sintassi | Modifica la lista | Restituisce |
|---|---|---|
| `lista.append(x)` | sì | `None` |

**Output atteso:** `['pane', 'latte', 'uova']`, `3` e `None`.

### Versione semplificata

## 11. Il metodo `insert()`

`insert(i, x)` inserisce `x` **nella posizione `i`**. Gli elementi che si trovano da `i` in poi **slittano di un posto verso destra**.

|---|---|---|
| `lista.insert(i, x)` | sì | `None` |

```text
 prima:   pane   latte   uova

 spesa.insert(1, "mele")

 dopo:    pane   mele   latte   uova
 indice     0      1      2       3
```

````python
def main():  # Definisce il programma principale.
    spesa = ["pane", "latte", "uova"]  # Lista di tre prodotti.
    spesa.insert(1, "mele")  # "mele" va in posizione 1.
    print(spesa)  # latte e uova sono slittati a destra.
    spesa.insert(0, "riso")  # "riso" va in testa.
    print(spesa)  # Tutti gli altri elementi sono slittati.
    print(spesa[2])  # In posizione 2 si trova ora "mele".

if __name__ == '__main__':  # Controlla l'avvio diretto.
    main()  # Avvia il programma.
````

## 11.1 `insert()` in fondo e con indice oltre la fine

Con indice uguale a `len(lista)` l'elemento va in fondo, come con `append()`. Anche con un indice **oltre la fine** l'elemento viene messo in fondo, senza errori.

**Output atteso:** `[1, 2, 3, 4]` e poi `[1, 2, 3, 4, 5]`.

````python
def main():  # Definisce il programma principale.
    numeri = [1, 2, 3]  # Lista di tre interi.
    numeri.insert(len(numeri), 4)  # Indice 3: posizione subito dopo l'ultimo elemento.
    print(numeri)  # 4 è in fondo.
    numeri.insert(100, 5)  # Indice oltre la fine: va comunque in fondo.
    print(numeri)  # 5 è in fondo.

if __name__ == '__main__':  # Controlla l'avvio diretto.
    main()  # Avvia il programma.
````

## 11.2 Confronto tra `append()` e `insert()`

Partendo dalla stessa lista vuota, `append()` aggiunge sempre in fondo, mentre `insert(0, x)` aggiunge sempre in testa. Lo stesso ciclo produce quindi due liste in **ordine opposto**.

**Output atteso:** `[1, 2, 3, 4]` e `[4, 3, 2, 1]`.

````python
def main():  # Definisce il programma principale.
    in_fondo = []  # Lista vuota per append.
    in_testa = []  # Lista vuota per insert.
    for n in range(1, 5):  # n assume 1, 2, 3, 4.
        in_fondo.append(n)  # Ogni numero va dopo gli altri.
        in_testa.insert(0, n)  # Ogni numero va prima degli altri.
    print(in_fondo)  # Ordine crescente.
    print(in_testa)  # Ordine decrescente.

if __name__ == '__main__':  # Controlla l'avvio diretto.
    main()  # Avvia il programma.
````

## Esercizio 1 · La media dei voti

**Consegna.** Leggere `n` voti interi (n ≥ 1) **memorizzandoli in una lista**. Mostrare la lista, la **media**, il voto **più alto**, il **più basso** e il numero di **sufficienze** (voti ≥ 6).

**Caso di prova.** Con n = 4 e voti 7, 5, 8, 6 l'output atteso è:

```text
Voti: [7, 5, 8, 6] - Media: 6.5
Max: 8 - Min: 5 - Sufficienze: 3
```

````python
def main():  # Definisce il programma principale.
    n = int(input("Numero di voti: "))  # Viene letto il numero di voti.
    voti = []  # Lista vuota, prima del ciclo.
    for i in range(n):  # Il ciclo viene ripetuto n volte.
        voto = int(input("Voto: "))  # Viene letto e convertito un voto.
        voti.append(voto)  # Il voto viene aggiunto in fondo.
    media = sum(voti) / len(voti)  # Somma divisa per il numero di voti.
    sufficienze = 0  # Il contatore parte da 0.
    for voto in voti:  # Viene percorsa la lista per valore.
        if voto >= 6:  # Il voto è sufficiente.
            sufficienze = sufficienze + 1  # Il contatore aumenta di 1.
    print("Voti:", voti, "- Media:", media)  # Vengono mostrati lista e media.
    print("Max:", max(voti), "- Min:", min(voti),
          "- Sufficienze:", sufficienze)  # Vengono mostrati massimo, minimo e sufficienze.

if __name__ == '__main__':  # Controlla l'avvio diretto.
    main()  # Avvia il programma.
````

**Spiegazione.**

- **Riempimento.** `voti = []` sta **prima** del ciclo; ogni `append()` aggiunge in fondo il voto appena letto. Dopo il ciclo la lista è `[7, 5, 8, 6]`.
- **Media.** `sum(voti) / len(voti)` calcola 26 / 4 = 6.5. Poiché n ≥ 1, la lista non è mai vuota e la divisione è sicura.
- **Massimo e minimo.** `max()` e `min()` restituiscono 8 e 5 senza bisogno di un ciclo.
- **Sufficienze.** Il secondo ciclo percorre la lista per valore; il contatore aumenta per 7, 8 e 6.

| Voto | `voto >= 6` | `sufficienze` |
|---|---|---|
| 7 | `True` | 1 |
| 5 | `False` | 1 |
| 8 | `True` | 2 |
| 6 | `True` | 3 |
