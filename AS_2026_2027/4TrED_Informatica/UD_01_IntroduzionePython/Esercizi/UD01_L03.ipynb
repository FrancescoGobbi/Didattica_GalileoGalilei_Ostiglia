## 4. Una chiamata, due riferimenti e ritorno

Durante la chiamata, `main()` attende. Dopo il ritorno, `valore` indica ancora `7`.


### Versione semplificata


```python
def mostra(numero):  # Riceve il dato dal chiamante.
    print("Dentro mostra:", numero)  # Mostra il parametro.
def main():  # Definisce il chiamante.
    valore = 7  # Assegna il valore iniziale.
    print("Prima:", valore)  # Mostra il valore prima della chiamata.
    mostra(valore)  # Esegue la funzione e attende il ritorno.
    print("Dopo:", valore)  # Mostra il valore rimasto invariato.
if __name__ == '__main__':  # Controlla l'avvio diretto.
    main()  # Avvia il programma.
```

### Versione con ID


```python
def mostra(numero):  # Definisce la funzione chiamata, con un proprio parametro locale.
    print('Dentro mostra: numero =', numero)  # Mostra il valore.
    print('ID di numero:', id(numero))  # Mostra l'ID.
def main():  # Definisce il chiamante con il proprio insieme di nomi locali.
    valore = 7  # Associa valore all'oggetto intero 7.
    print('Prima della chiamata:', valore)  # Mostra il valore.
    print('ID di valore:', id(valore))  # Mostra l'ID.
    mostra(valore)  # Attiva mostra e attende che il suo corpo termini.
    print('Dopo il ritorno:', valore)  # Mostra il valore.
    print('ID di valore:', id(valore))  # Mostra l'ID.
if __name__ == '__main__':  # Verifica la condizione di avvio del programma.
    main()  # Esegue il programma principale.
```

### Seguire lo stack

Prima: è attiva `main()`. Durante la chiamata: è attiva anche `mostra()`. Al ritorno: `main()` riprende.


## 5. Riassegnare un parametro

**Output:** `8`, poi `7`. La riassegnazione modifica soltanto il riferimento locale `numero`.


### Versione semplificata


```python
def aumenta(numero):  # Riceve il riferimento al dato.
    numero = numero + 1  # Riassegna soltanto il parametro locale.
    print(numero)  # Mostra 8.
def main():  # Definisce il chiamante.
    valore = 7  # Assegna il valore iniziale.
    aumenta(valore)  # Esegue la funzione.
    print(valore)  # Mostra ancora 7.
if __name__ == '__main__':  # Controlla l'avvio diretto.
    main()  # Avvia il programma.
```

### Versione con ID


```python
def aumenta(numero):  # Riceve un riferimento all'oggetto indicato dall'argomento.
    print('Parametro prima:', numero)  # Mostra il valore.
    print('ID di numero:', id(numero))  # Mostra l'ID.
    numero = numero + 1  # Associa il parametro locale al risultato 8, senza modificare l'intero 7.
    print('Parametro dopo:', numero)  # Mostra il valore.
    print('ID di numero:', id(numero))  # Mostra l'ID.
def main():  # Definisce il programma chiamante.
    valore = 7  # Associa valore all'intero iniziale.
    print('Chiamante prima:', valore)  # Mostra il valore.
    print('ID di valore:', id(valore))  # Mostra l'ID.
    aumenta(valore)  # Passa il riferimento senza consentire la riassegnazione del nome valore dall'interno della funzione.
    print('Chiamante dopo:', valore)  # Mostra il valore.
    print('ID di valore:', id(valore))  # Mostra l'ID.
if __name__ == '__main__':  # Controlla l'avvio del programma principale.
    main()  # Avvia l'esempio.
```

```python
def doppio(x):  # Riceve il numero.
    return 2 * x  # Restituisce il doppio.
def main():  # Definisce il chiamante.
    risultato = doppio(3)  # Riceve il risultato.
    print(risultato)  # Mostra 6.
if __name__ == '__main__':  # Controlla l'avvio diretto.
    main()  # Avvia il programma.
```

## 8. Funzione sommatoria con for

**Consegna:** sommare da 1 a `n` con `for`, per `n` intero non negativo.

**Prove:** `4 → 10`, `1 → 1`, `0 → 0`, `5 → 15`.


### Versione semplificata


```python
def sommatoria(n):  # Riceve un intero non negativo.
    totale = 0  # Inizializza la somma.
    for numero in range(1, n + 1):  # Include gli interi da 1 a n.
        totale = totale + numero  # Aggiunge il valore corrente.
    return totale  # Restituisce la somma completa.
def main():  # Esegue i casi di prova.
    for valore in (4, 1, 0, 5):  # Prova i quattro argomenti.
        somma = sommatoria(valore)  # Calcola la somma.
        print("Sommatoria di", valore, "=", somma)  # Mostra il risultato.
if __name__ == '__main__':  # Controlla l'avvio diretto.
    main()  # Avvia le prove.
```

### Versione con ID


```python
def sommatoria(n):  # Riceve un intero non negativo, assumendo valido l'argomento.
    print('Parametro n:', n)  # Mostra il valore.
    print('ID di n:', id(n))  # Mostra l'ID.
    totale = 0  # Inizializza l'accumulatore della somma per questa chiamata.
    print('Totale iniziale:', totale)  # Mostra il valore.
    print('ID di totale:', id(totale))  # Mostra l'ID.
    for numero in range(1, n + 1):  # Percorre gli interi da 1 a n inclusi.
        print('Prima: totale =', totale)  # Mostra il valore.
        print('ID di totale:', id(totale))  # Mostra l'ID.
        totale = totale + numero  # Associa totale al risultato della somma corrente.
        print('Addendo:', numero, 'Totale dopo:', totale)  # Mostra il valore.
        print('ID di totale:', id(totale))  # Mostra l'ID.
    print('Totale restituito:', totale)  # Mostra il valore.
    print('ID di totale:', id(totale))  # Mostra l'ID.
    return totale  # Restituisce il totale completo, fuori dal ciclo.
def main():  # Raccoglie tutte le prove presenti nelle slide per questa funzione.
    for valore in (4, 1, 0, 5):  # Esegue separatamente i quattro casi indicati, usando una tupla di valori di prova.
        print('Chiamata sommatoria con', valore)  # Mostra il valore.
        print('ID di valore:', id(valore))  # Mostra l'ID.
        somma = sommatoria(valore)  # Riceve il risultato della nuova chiamata.
        print('Somma:', somma)  # Mostra il valore.
        print('ID di somma:', id(somma))  # Mostra l'ID.
if __name__ == '__main__':  # Controlla l'avvio dell'esempio.
    main()  # Avvia i casi di prova.
```

## 9. Funzione fattoriale con while

**Consegna:** calcolare `n!` con `while`, per `n` intero non negativo. Il prodotto parte da 1.

**Prove:** `4 → 24`, `5 → 120`, `0 → 1`.


### Versione semplificata


```python
def fattoriale(n):  # Riceve un intero non negativo.
    prodotto = 1  # Inizializza il prodotto.
    fattore = 1  # Imposta il primo fattore.
    while fattore <= n:  # Include il fattore n.
        prodotto = prodotto * fattore  # Aggiorna il prodotto.
        fattore = fattore + 1  # Passa al fattore successivo.
    return prodotto  # Restituisce il fattoriale.
def main():  # Esegue i casi di prova.
    for valore in (4, 5, 0):  # Prova anche il caso zero.
        risultato = fattoriale(valore)  # Calcola il fattoriale.
        print("Fattoriale di", valore, "=", risultato)  # Mostra il risultato.
if __name__ == '__main__':  # Controlla l'avvio diretto.
    main()  # Avvia le prove.
```

### Versione con ID


```python
def fattoriale(n):  # Riceve l'intero non negativo di cui calcolare il fattoriale.
    print('Parametro n:', n)  # Mostra il valore.
    print('ID di n:', id(n))  # Mostra l'ID.
    prodotto = 1  # Inizializza il prodotto all'elemento neutro della moltiplicazione.
    fattore = 1  # Associa il contatore al primo fattore da usare.
    print('Prodotto iniziale:', prodotto)  # Mostra il valore.
    print('ID di prodotto:', id(prodotto))  # Mostra l'ID.
    while fattore <= n:  # Ripete il corpo fino a includere il fattore n.
        print('Prima: prodotto =', prodotto)  # Mostra il valore.
        print('ID di prodotto:', id(prodotto))  # Mostra l'ID.
        prodotto = prodotto * fattore  # Associa l'accumulatore al risultato della moltiplicazione.
        print('Fattore:', fattore, 'Prodotto dopo:', prodotto)  # Mostra il valore.
        print('ID di prodotto:', id(prodotto))  # Mostra l'ID.
        fattore = fattore + 1  # Prepara il fattore successivo.
    print('Prodotto restituito:', prodotto)  # Mostra il valore.
    print('ID di prodotto:', id(prodotto))  # Mostra l'ID.
    return prodotto  # Restituisce il risultato dopo tutte le iterazioni.
def main():  # Definisce le prove della funzione fattoriale.
    for valore in (4, 5, 0):  # Ripete la chiamata per tutti i casi presenti nelle slide.
        print('Chiamata fattoriale con', valore)  # Mostra il valore.
        print('ID di valore:', id(valore))  # Mostra l'ID.
        risultato = fattoriale(valore)  # Riceve il prodotto calcolato dalla funzione.
        print('Fattoriale:', risultato)  # Mostra il valore.
        print('ID di risultato:', id(risultato))  # Mostra l'ID.
if __name__ == '__main__':  # Verifica l'avvio del programma principale.
    main()  # Esegue i casi di prova.
```

## 10. Il maggiore di due numeri

**Consegna:** restituire il maggiore tra due numeri; in caso di parità, restituire il primo.

**Prove:** `(8, 5) → 8`, `(-2, -7) → -2`, `(3, 3) → 3`, `(2, 9) → 9`.


### Versione semplificata


```python
def maggiore(a, b):  # Riceve due numeri.
    if a >= b:  # Include il caso di parità.
        return a  # Restituisce il primo argomento.
    return b  # Restituisce il secondo se è maggiore.
def main():  # Esegue i casi di prova.
    for primo, secondo in ((8, 5), (-2, -7), (3, 3), (2, 9)):  # Percorre le coppie.
        massimo = maggiore(primo, secondo)  # Trova il maggiore.
        print("Maggiore tra", primo, "e", secondo, "=", massimo)  # Mostra il risultato.
if __name__ == '__main__':  # Controlla l'avvio diretto.
    main()  # Avvia le prove.
```

### Versione con ID


```python
def maggiore(a, b):  # Riceve due riferimenti in ordine posizionale.
    print('Parametro a:', a)  # Mostra il valore.
    print('ID di a:', id(a))  # Mostra l'ID.
    print('Parametro b:', b)  # Mostra il valore.
    print('ID di b:', id(b))  # Mostra l'ID.
    if a >= b:  # Seleziona il primo argomento anche in caso di parità.
        print('ID di a:', id(a))  # Mostra l'ID.
        return a  # Termina la funzione restituendo lo stesso oggetto indicato da a.
    print('ID di b:', id(b))  # Mostra l'ID.
    return b  # Termina la funzione restituendo lo stesso oggetto indicato da b.
def main():  # Definisce il chiamante con i casi di prova.
    for primo, secondo in ((8, 5), (-2, -7), (3, 3), (2, 9)):  # Assegna ai due nomi ciascuna coppia di prova.
        print('Primo:', primo)  # Mostra il valore.
        print('ID di primo:', id(primo))  # Mostra l'ID.
        print('Secondo:', secondo)  # Mostra il valore.
        print('ID di secondo:', id(secondo))  # Mostra l'ID.
        massimo = maggiore(primo, secondo)  # Associa massimo all'oggetto selezionato dalla funzione.
        print('Massimo:', massimo)  # Mostra il valore.
        print('ID di massimo:', id(massimo))  # Mostra l'ID.
        print("Stesso oggetto del primo:", massimo is primo)  # Mostra il valore.
        print("Stesso oggetto del secondo:", massimo is secondo)  # Mostra il valore.
if __name__ == '__main__':  # Controlla l'avvio del programma.
    main()  # Esegue tutte le prove.
```

### Valori e ritorno

`==` confronta i valori. `return a` restituisce l’oggetto indicato da `a`, senza copiarlo.


## 11. Input nel programma principale

**Consegna:** leggere un intero in `main()`, passarlo a `doppio()` e stampare il risultato.

**Prove:** `5 → 10`, `0 → 0`.


### Versione semplificata


```python
def doppio(x):  # Riceve il numero già acquisito.
    return 2 * x  # Restituisce il doppio.
def main():  # Gestisce input e output.
    numero = int(input("Numero: "))  # Legge e converte l'intero.
    risultato = doppio(numero)  # Calcola il doppio.
    print("Doppio:", risultato)  # Mostra il risultato.
if __name__ == '__main__':  # Controlla l'avvio diretto.
    main()  # Avvia il programma.
```

### Versione con ID


```python
def doppio(x):  # Definisce il calcolo riutilizzabile senza acquisire input.
    print('Parametro x:', x)  # Mostra il valore.
    print('ID di x:', id(x))  # Mostra l'ID.
    calcolato = 2 * x  # Associa il nome locale al risultato del calcolo.
    print('Risultato nella funzione:', calcolato)  # Mostra il valore.
    print('ID di calcolato:', id(calcolato))  # Mostra l'ID.
    return calcolato  # Restituisce il riferimento al risultato.
def main():  # Gestisce ingresso, chiamata e uscita.
    numero = int(input("Numero: "))  # Legge una stringa e associa numero al risultato della conversione in intero.
    print('Numero in main:', numero)  # Mostra il valore.
    print('ID di numero:', id(numero))  # Mostra l'ID.
    risultato = doppio(numero)  # Associa risultato all'oggetto restituito dalla funzione.
    print("Doppio:", risultato)  # Mostra il valore.
    print('ID di risultato:', id(risultato))  # Mostra l'ID.
    print('Numero dopo la chiamata:', numero)  # Mostra il valore.
    print('ID di numero:', id(numero))  # Mostra l'ID.
if __name__ == '__main__':  # Verifica l'avvio del programma principale.
    main()  # Avvia la lettura del dato e il calcolo.
```
