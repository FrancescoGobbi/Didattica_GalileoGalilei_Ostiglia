## Come lavorare nel notebook

Gli esempi riprendono quelli delle slide. Ogni esempio principale ha due versioni:

- **versione semplificata**: solo il codice necessario e l'output essenziale;
- **versione con stampe di verifica**: mostra ogni chiamata, il valore di `n`, il suo ID e il valore restituito.

Come leggere le stampe di verifica:

- ogni chiamata annidata sposta le righe di quattro spazi verso destra;
- più una riga è rientrata, più in alto si trova il suo frame nello stack.

**Nota bene.** Gli ID possono cambiare da un'esecuzione all'altra. Python può riutilizzare gli oggetti dei piccoli interi.

## La ricorsione in sintesi

| Concetto | Significato | Esempio con la sommatoria |
|---|---|---|
| Funzione ricorsiva | Richiama se stessa | `sommatoria` chiama `sommatoria` |
| Caso base | Si risolve subito, senza nuove chiamate | con `n == 0` il risultato è `0` |
| Caso ricorsivo | Usa lo stesso problema su un dato più piccolo | `n + sommatoria(n - 1)` |
| Avvicinamento al caso base | Ogni chiamata riceve un dato più vicino al caso base | 3, 2, 1, 0 |
| Frame | Spazio di una chiamata: contiene le etichette locali e il punto da cui riprendere | `n` indica 3 in un frame e 2 in un altro |
| Stack delle chiamate | Pila dei frame attivi: l'ultimo creato è il primo a terminare | `main()` → `sommatoria(3)` → `sommatoria(2)` → … |

## Lo schema di una funzione ricorsiva

Questo schema contiene segnaposto: è solo testo, non codice eseguibile.

```text
def funzione(dato):
    if caso base:                                  # Il risultato è noto senza nuove chiamate.
        return risultato noto
    else:                                          # Caso ricorsivo.
        return operazione con funzione(dato più piccolo)
```

**Nota bene.**

- Il caso base si trova nel ramo `if`.
- Il caso ricorsivo si trova nel ramo `else`.
- Il dato passato alla nuova chiamata deve avvicinarsi al caso base, altrimenti la ricorsione non termina.

## 1. La sommatoria ricorsiva

La funzione calcola la somma degli interi da 1 a `n`, con `n` intero non negativo.

- Caso base: `S(0) = 0`
- Caso ricorsivo: `S(n) = n + S(n - 1)`

**Prove:** `3 → 6`, `0 → 0`, `4 → 10`, `5 → 15`.

### Versione semplificata

````python
def sommatoria(n):  # Riceve un intero non negativo.
    if n == 0:  # Caso base.
        return 0  # La somma senza addendi vale 0.
    else:  # Caso ricorsivo.
        return n + sommatoria(n - 1)  # Viene aggiunto n alla somma fino a n - 1.

def main():  # Esegue i casi di prova.
    for valore in (3, 0, 4, 5):  # Vengono provati quattro argomenti.
        print("Sommatoria di", valore, "=", sommatoria(valore))  # Viene mostrato il risultato.

if __name__ == '__main__':  # Controlla l'avvio diretto.
    main()  # Avvia le prove.
````

### Versione con stampe di verifica

Rispetto alla versione semplificata ci sono due aggiunte:

- il parametro `livello` conta le chiamate di `sommatoria` già aperte e serve solo per il rientro delle righe;
- l'etichetta `parziale` indica il risultato della chiamata interna e serve solo per mostrarlo.

L'espressione `'    ' * livello` ripete quattro spazi tante volte quanto vale `livello`.

Il calcolo è lo stesso della versione semplificata.

````python
def sommatoria(n, livello):  # Riceve n e il numero di chiamate già aperte.
    rientro = '    ' * livello  # Viene preparato il rientro della riga.
    print(f"{rientro}Chiamata sommatoria({n}) · ID di n: {id(n)}")  # Viene mostrata la nuova chiamata.
    if n == 0:  # Caso base.
        print(f"{rientro}Caso base: restituisce 0")  # Il risultato è noto.
        return 0  # La chiamata termina senza crearne altre.
    else:  # Caso ricorsivo.
        print(f"{rientro}In sospeso: {n} + ?")  # La somma attende il risultato della chiamata interna.
        parziale = sommatoria(n - 1, livello + 1)  # Viene creata una nuova chiamata, un livello più in alto.
        totale = n + parziale  # Al ritorno la somma viene completata.
        print(f"{rientro}Ritorno: {n} + {parziale} = {totale} · ID di n: {id(n)}")  # n indica ancora lo stesso oggetto.
        return totale  # Il risultato viene restituito al chiamante.

def main():  # Avvia la versione con stampe di verifica.
    risultato = sommatoria(3, 0)  # La prima chiamata parte dal livello 0.
    print("Risultato in main:", risultato)  # Viene mostrato 6.

if __name__ == '__main__':  # Controlla l'avvio diretto.
    main()  # Avvia il programma.
````

### Seguire lo stack

| Chiamata | Valore di `n` | Operazione in sospeso | Valore restituito |
|---|---|---|---|
| `sommatoria(3)` | 3 | `3 + ?` | 6 |
| `sommatoria(2)` | 2 | `2 + ?` | 3 |
| `sommatoria(1)` | 1 | `1 + ?` | 1 |
| `sommatoria(0)` | 0 | nessuna: caso base | 0 |

Come leggere la tabella:

- le chiamate vengono create dall'alto verso il basso;
- i ritorni avvengono dal basso verso l'alto;
- `3 + ?` viene completata solo quando `sommatoria(2)` ha restituito il proprio risultato.

Nelle stampe di verifica, l'ID di `n` è uguale alla chiamata e al ritorno dello stesso frame: nessuna chiamata interna modifica l'etichetta `n` del chiamante.

## 2. Prima l'argomento, poi `print()`

In `print("Risultato:", sommatoria(3))` Python calcola prima gli argomenti e solo dopo chiama `print()`.

Durante la ricorsione, quindi, `print()` non è ancora nello stack. Entra sopra `main()` solo quando `sommatoria(3)` ha restituito 6.

**Output atteso**

```text
Calcolo con n = 3
Calcolo con n = 2
Calcolo con n = 1
Caso base raggiunto
Risultato: 6
```

````python
def sommatoria(n):  # Riceve un intero non negativo.
    if n == 0:  # Caso base.
        print("Caso base raggiunto")  # Viene segnalato l'arrivo al caso base.
        return 0  # Viene restituito il valore noto.
    else:  # Caso ricorsivo.
        print("Calcolo con n =", n)  # Viene segnalata la chiamata corrente.
        return n + sommatoria(n - 1)  # La somma attende il risultato interno.

def main():  # Definisce il programma principale.
    print("Risultato:", sommatoria(3))  # print viene chiamata solo dopo il calcolo dell'argomento.

if __name__ == '__main__':  # Controlla l'avvio diretto.
    main()  # Avvia il programma.
````

## 3. Il fattoriale ricorsivo

La funzione calcola `n!`, con `n` intero non negativo.

- Caso base: `0! = 1`
- Caso ricorsivo: `n! = n · (n - 1)!`

Rispetto alla sommatoria cambiano il valore del caso base e l'operazione in sospeso.

**Prove:** `3 → 6`, `4 → 24`, `0 → 1`, `5 → 120`.

### Versione semplificata

````python
def fattoriale(n):  # Riceve un intero non negativo.
    if n == 0:  # Caso base.
        return 1  # Per definizione 0! = 1.
    else:  # Caso ricorsivo.
        return n * fattoriale(n - 1)  # Viene moltiplicato n per il fattoriale di n - 1.

def main():  # Esegue i casi di prova.
    for valore in (3, 4, 0, 5):  # Vengono provati quattro argomenti.
        print("Fattoriale di", valore, "=", fattoriale(valore))  # Viene mostrato il risultato.

if __name__ == '__main__':  # Controlla l'avvio diretto.
    main()  # Avvia le prove.
````

### Versione con stampe di verifica

Le stampe di verifica hanno la stessa struttura di quelle della sommatoria:

- `livello` serve per il rientro delle righe;
- `parziale` serve per mostrare il valore restituito dalla chiamata interna.

````python
def fattoriale(n, livello):  # Riceve n e il numero di chiamate già aperte.
    rientro = '    ' * livello  # Viene preparato il rientro della riga.
    print(f"{rientro}Chiamata fattoriale({n}) · ID di n: {id(n)}")  # Viene mostrata la nuova chiamata.
    if n == 0:  # Caso base.
        print(f"{rientro}Caso base: restituisce 1")  # Il risultato è noto.
        return 1  # La chiamata termina senza crearne altre.
    else:  # Caso ricorsivo.
        print(f"{rientro}In sospeso: {n} * ?")  # Il prodotto attende il risultato interno.
        parziale = fattoriale(n - 1, livello + 1)  # Viene creata una nuova chiamata.
        prodotto = n * parziale  # Al ritorno il prodotto viene completato.
        print(f"{rientro}Ritorno: {n} * {parziale} = {prodotto} · ID di n: {id(n)}")  # Viene mostrato il risultato parziale.
        return prodotto  # Il risultato viene restituito al chiamante.

def main():  # Avvia la versione con stampe di verifica.
    risultato = fattoriale(3, 0)  # La prima chiamata parte dal livello 0.
    print("Risultato in main:", risultato)  # Viene mostrato 6.

if __name__ == '__main__':  # Controlla l'avvio diretto.
    main()  # Avvia il programma.
````

### Seguire lo stack

| Chiamata | Valore di `n` | Operazione in sospeso | Valore restituito |
|---|---|---|---|
| `fattoriale(3)` | 3 | `3 * ?` | 6 |
| `fattoriale(2)` | 2 | `2 * ?` | 2 |
| `fattoriale(1)` | 1 | `1 * ?` | 1 |
| `fattoriale(0)` | 0 | nessuna: caso base | 1 |

Al caso base lo stack contiene quattro frame di `fattoriale` sopra `main()`.

Le moltiplicazioni avvengono solo durante la risalita:

1. `1 * 1`
2. `2 * 1`
3. `3 * 2`

## 4. Ogni chiamata conserva la propria etichetta `n`

La funzione `conta()` non restituisce un valore: qui conta l'**ordine delle stampe**. Ogni frame mostra `n` prima della chiamata interna e di nuovo dopo il ritorno.

**Output atteso**

```text
Prima: n = 3
Prima: n = 2
Prima: n = 1
Caso base
Dopo: n = 1
Dopo: n = 2
Dopo: n = 3
```

Le stampe `Dopo` compaiono in ordine inverso, perché l'ultimo frame creato è il primo a terminare.

### Versione semplificata

````python
def conta(n):  # Riceve un intero non negativo.
    if n == 0:  # Caso base.
        print("Caso base")  # Non vengono create altre chiamate.
    else:  # Caso ricorsivo.
        print("Prima: n =", n)  # Viene mostrato n prima della chiamata interna.
        conta(n - 1)  # Il frame corrente resta in attesa.
        print("Dopo: n =", n)  # Al ritorno n indica ancora lo stesso valore.

def main():  # Definisce il programma principale.
    conta(3)  # Avvia la ricorsione con n = 3.

if __name__ == '__main__':  # Controlla l'avvio diretto.
    main()  # Avvia il programma.
````

### Versione con ID

````python
def conta(n):  # Riceve un intero non negativo.
    if n == 0:  # Caso base.
        print("Caso base · ID di n:", id(n))  # Viene mostrato l'ID dell'oggetto 0.
    else:  # Caso ricorsivo.
        print("Prima: n =", n, "· ID di n:", id(n))  # Viene mostrato l'ID prima della chiamata.
        conta(n - 1)  # La nuova chiamata associa la propria etichetta n a un altro oggetto.
        print("Dopo:  n =", n, "· ID di n:", id(n))  # L'ID coincide con quello mostrato prima.

def main():  # Definisce il programma principale.
    conta(3)  # Avvia la ricorsione con n = 3.

if __name__ == '__main__':  # Controlla l'avvio diretto.
    main()  # Avvia il programma.
````

## 5. Stampare non significa restituire

Nel caso base c'è `print(0)` al posto di `return 0`.

- Il valore 0 compare sullo schermo, ma la chiamata restituisce `None`.
- Nel frame con `n = 1` l'operazione in sospeso diventa `1 + None`.
- Questa operazione non può essere calcolata.

**Nota bene.** La cella produce volutamente un errore: compare `0`, poi un messaggio che termina con `TypeError: unsupported operand type(s) for +: 'int' and 'NoneType'`.

````python
def sommatoria_errata(n):  # Versione con un errore nel caso base.
    if n == 0:  # Caso base.
        print(0)  # Errore: il valore viene mostrato, non restituito.
    else:  # Caso ricorsivo.
        return n + sommatoria_errata(n - 1)  # Riceve None dalla chiamata con n = 0.

def main():  # Definisce il programma principale.
    print(sommatoria_errata(2))  # L'esecuzione si interrompe con TypeError.

if __name__ == '__main__':  # Controlla l'avvio diretto.
    main()  # Avvia il programma.
````

## 6. Ricorsione senza fine e `RecursionError`

Con un argomento negativo il test `n == 0` non diventa mai vero.

- Gli argomenti sono −1, −2, −3, … e si allontanano dal caso base.
- Ogni chiamata aggiunge un frame allo stack.
- A un certo punto viene superato il limite di profondità fissato da Python.

Il modulo `sys` contiene informazioni sull'interprete. `sys.getrecursionlimit()` restituisce il limite, di norma 1000.

**Nota bene.** La cella produce volutamente un errore: il messaggio finale inizia con `RecursionError: maximum recursion depth exceeded`.

````python
import sys  # Viene importato il modulo con le informazioni sull'interprete.

def sommatoria(n):  # Funzione corretta solo per interi non negativi.
    if n == 0:  # Caso base, mai raggiunto con argomenti negativi.
        return 0  # Viene restituito il valore noto.
    else:  # Caso ricorsivo.
        return n + sommatoria(n - 1)  # Con n negativo l'argomento si allontana da 0.

def main():  # Definisce il programma principale.
    print("Limite di profondità:", sys.getrecursionlimit())  # Viene mostrato il limite.
    print(sommatoria(-1))  # La precondizione n >= 0 non è rispettata.

if __name__ == '__main__':  # Controlla l'avvio diretto.
    main()  # Avvia il programma.
````

## 7. Ricorsione e iterazione a confronto

Qui la sommatoria della lezione 03, scritta con `for`, viene confrontata con la versione ricorsiva. Le due funzioni restituiscono gli stessi risultati.

| Aspetto | Versione con `for` | Versione ricorsiva |
|---|---|---|
| Ripetizione | Il ciclo ripete il corpo | La funzione richiama se stessa |
| Stato del calcolo | L'accumulatore `totale` viene aggiornato | Ogni frame conserva la propria `n` e l'operazione in sospeso |
| Arresto | I valori del `range()` terminano | Viene raggiunto il caso base |
| Frame nello stack | Uno solo per tutta la chiamata | `n + 1` frame per `sommatoria(n)` |

**Prove:** `4 → 10`, `1 → 1`, `0 → 0`, `5 → 15`.

````python
def sommatoria_for(n):  # Versione iterativa della lezione 03.
    totale = 0  # Viene inizializzato l'accumulatore.
    for numero in range(1, n + 1):  # Vengono percorsi gli interi da 1 a n.
        totale = totale + numero  # Viene aggiunto il valore corrente.
    return totale  # Viene restituita la somma completa.

def sommatoria_ricorsiva(n):  # Versione ricorsiva.
    if n == 0:  # Caso base.
        return 0  # Viene restituito il valore noto.
    else:  # Caso ricorsivo.
        return n + sommatoria_ricorsiva(n - 1)  # Viene usato il risultato della chiamata interna.

def main():  # Confronta le due versioni sugli stessi casi di prova.
    for valore in (4, 1, 0, 5):  # Vengono provati quattro argomenti.
        print("n =", valore,
              "· for:", sommatoria_for(valore),
              "· ricorsiva:", sommatoria_ricorsiva(valore))  # I due risultati coincidono.

if __name__ == '__main__':  # Controlla l'avvio diretto.
    main()  # Avvia il confronto.
````

## Esercizio svolto 1 · Il numero di caselle

**Consegna.** Una figura ha `n` righe: la prima contiene 1 casella, la seconda 2 e così via fino alla riga `n`.

Creare una funzione ricorsiva `caselle(n)` che restituisca il numero totale di caselle, con `n` intero non negativo. Una figura senza righe contiene 0 caselle.

```text
□          riga 1
□□         riga 2
□□□        riga 3
□□□□       riga 4: ultima riga, con 4 caselle
```

**Casi di prova**

- `n = 4` → 10
- `n = 1` → 1
- `n = 0` → 0

````python
def caselle(n):  # Riceve il numero di righe della figura.
    if n == 0:  # Caso base: figura senza righe.
        return 0  # Non ci sono caselle.
    else:  # Caso ricorsivo.
        return n + caselle(n - 1)  # Ultima riga più la figura con una riga in meno.

def main():  # Esegue i casi di prova.
    for righe in (4, 1, 0):  # Vengono provati tre argomenti.
        print("Righe:", righe, "· Caselle:", caselle(righe))  # Viene mostrato il risultato.

if __name__ == '__main__':  # Controlla l'avvio diretto.
    main()  # Avvia le prove.
````

**Spiegazione della soluzione.** La figura con `n` righe è formata da:

- l'ultima riga, che contiene `n` caselle;
- una figura più piccola, con `n - 1` righe.

Il problema si riduce allo stesso problema su un dato più piccolo, fino alla figura vuota.

Con `n = 4`:

- gli argomenti sono 4, 3, 2, 1, 0;
- i valori restituiti, dal caso base alla prima chiamata, sono 0, 1, 3, 6, 10.

## Esercizio svolto 2 · Una potenza intera

**Consegna.** Creare una funzione ricorsiva `potenza(b, e)` che calcoli `b` elevato a `e`, con `b` intero maggiore di zero ed `e` intero non negativo.

Usare solo moltiplicazioni, **senza l'operatore** `**`. Con esponente 0 il risultato è 1.

**Casi di prova**

- `(2, 3)` → 8
- `(5, 0)` → 1
- `(3, 4)` → 81

````python
def potenza(b, e):  # Riceve la base e l'esponente.
    if e == 0:  # Caso base.
        return 1  # Qualsiasi base elevata a zero vale 1.
    else:  # Caso ricorsivo.
        return b * potenza(b, e - 1)  # La base resta uguale; l'esponente diminuisce di 1.

def main():  # Esegue i casi di prova.
    for base, esponente in ((2, 3), (5, 0), (3, 4)):  # Vengono provate tre coppie.
        print(base, "elevato a", esponente, "=", potenza(base, esponente))  # Viene mostrato il risultato.

if __name__ == '__main__':  # Controlla l'avvio diretto.
    main()  # Avvia le prove.
````

**Spiegazione della soluzione.** Elevare `b` a `e` significa moltiplicare `b` per `b` elevato a `e - 1`.

Ogni chiamata riceve la stessa base e un esponente ridotto di 1: è l'esponente ad avvicinarsi al caso base `e == 0`.

| Chiamata | Operazione in sospeso | Valore restituito |
|---|---|---|
| `potenza(2, 3)` | `2 * ?` | 8 |
| `potenza(2, 2)` | `2 * ?` | 4 |
| `potenza(2, 1)` | `2 * ?` | 2 |
| `potenza(2, 0)` | nessuna: caso base | 1 |

## Esercizio 1 · Conto alla rovescia ricorsivo

**Consegna.** Creare una funzione ricorsiva `conto_alla_rovescia(n)` che mostri i numeri da `n` a 1, uno per riga, e infine il messaggio `Partenza!`.

**Richiesta**

- **Non usare cicli.**
- Il caso base è `n == 0`: viene mostrato soltanto `Partenza!`.

**Casi di prova**

- `n = 3` → 3, 2, 1, `Partenza!`
- `n = 1` → 1, `Partenza!`
- `n = 0` → `Partenza!`

````python
# Codice per l'esercizio 01

def countdown(n):  # Riceve un intero non negativo
    if n == 0:  # Caso base
        return  # Non viene mostrato nulla
    else:  # Caso ricorsivo
        print(n)  # Viene mostrato il numero corrente
        countdown(n - 1)  # Viene creata una nuova chiamata con n diminuito di 1

def main():
    n = int(input("Inserisci un numero intero non negativo: "))  # Chiede all'utente un numero
    countdown(n)  # Avvia il conto alla rovescia
    print("Partenza!") # Viene mostrato il messaggio finale

# Parte il modulo principale
if __name__ == '__main__':
    main()
````

## Esercizio 2 · Somma dei quadrati

**Consegna.** Creare una funzione ricorsiva `somma_quadrati(n)` che restituisca `1² + 2² + … + n²`, con `n` intero non negativo.

**Richiesta**

- La funzione deve **restituire** il risultato con `return`.
- La stampa avviene in `main()`.
- In `main()` è ammesso un ciclo `for` per eseguire i casi di prova.

**Casi di prova**

- `n = 0` → 0
- `n = 1` → 1
- `n = 3` → 14
- `n = 4` → 30

````python
# Codice per l'esercizio 02

def somma_quadrati(n):  # Riceve un intero non negativo
    if n == 0:  # Caso base
        return 0  # La somma senza addendi vale 0
    else:  # Caso ricorsivo
        return n**2 + somma_quadrati(n - 1)  # Viene aggiunto il quadrato di n alla somma fino a n - 1

def main():  # Esegue i casi di prova
    n = int(input("Inserisci un numero intero non negativo: "))  # Chiede all'utente un numero
    risultato = somma_quadrati(n)  # Calcola la somma dei quadrati fino a n
    print(f"La somma dei quadrati da 1 a {n} è: {risultato}")  # Viene mostrato il risultato

# Parte il modulo principale
if __name__ == '__main__':
    main()
````

**Consegna.** Creare una funzione ricorsiva `conta_cifre(n)` che restituisca il numero di cifre di un intero `n` maggiore o uguale a 1.

**Richiesta**

- Un numero minore di 10 ha una sola cifra.
