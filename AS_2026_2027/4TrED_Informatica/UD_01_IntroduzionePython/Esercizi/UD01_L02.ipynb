## La sintassi dei cicli
Con `while condizione:` il corpo viene eseguito finché la condizione risulta vera; il controllo avviene prima di ogni iterazione. Nei conteggi, il contatore viene inizializzato prima del ciclo e aggiornato al suo interno.

Con `for nome in intervallo:` il nome assume automaticamente, uno alla volta, i valori dell'intervallo. Per le iterazioni numeriche si utilizza `range()`.

**Nota bene.** I due punti introducono il corpo. L'indentazione definisce il blocco di istruzioni ripetuto: negli esempi si usano quattro spazi. Un'istruzione successiva allineata al ciclo si trova fuori dal suo corpo.


## `while`: lo schema concettuale
Lo schema seguente descrive le parti del costrutto e viene riportato come testo, poiché contiene segnaposto.

```text
inizializzazione             # Viene stabilito il valore iniziale.
while condizione:            # La condizione viene verificata prima del corpo.
    istruzioni del corpo     # Vengono eseguite se la condizione è vera.
    aggiornamento            # Nel conteggio viene modificato il contatore.
istruzioni successive        # L'esecuzione prosegue quando il ciclo termina.
```


## 1. `while`: contare da 1 a 5

Il contatore parte da 1. Il corpo stampa il valore e lo incrementa; quando il valore raggiunge 6, la condizione diventa falsa. Vengono mostrati i numeri da 1 a 5 e poi `Fine`.

**Nota bene.** Stampa e aggiornamento appartengono al blocco indentato; la stampa di `Fine` è esterna al ciclo.


````python
numero = 1                 # Viene inizializzato il contatore.
while numero <= 5:         # Viene incluso anche il valore 5.
    print(numero)          # Viene mostrato il valore corrente.
    numero = numero + 1    # Il contatore avanza verso la condizione di arresto.
print("Fine")               # Il messaggio compare una sola volta dopo il ciclo.
````

### L'esecuzione del `while`
| Valore al controllo | Condizione | Valore stampato | Valore dopo l'aggiornamento |
|---|---|---|---|
| 1 | True | 1 | 2 |
| 2 | True | 2 | 3 |
| 3 | True | 3 | 4 |
| 4 | True | 4 | 5 |
| 5 | True | 5 | 6 |
| 6 | False | Nessuno | Il ciclo termina |

Il corpo viene eseguito cinque volte; la condizione viene controllata sei volte. Se la condizione è falsa all'inizio, il corpo non viene eseguito.


## 2. `while`: conto alla rovescia

Il contatore diminuisce a ogni passaggio. Il valore 1 viene incluso; dopo l’aggiornamento a 0 il ciclo termina e compare `Via!`.


````python
numero = 5                 # Il conto alla rovescia parte da 5.
while numero >= 1:         # Il corpo viene eseguito fino a 1 incluso.
    print(numero)          # Viene mostrato il valore corrente.
    numero = numero - 1    # Il contatore si avvicina allo zero.
print("Via!")               # Il messaggio viene mostrato dopo il ciclo.
````

## 3. Accumulare una somma con `while`

L’accumulatore `somma` conserva il totale parziale. Viene inizializzato prima del ciclo, affinché il totale non venga azzerato a ogni iterazione. La somma di 1, 2, 3 e 4 è 10.


````python
numero = 1                   # Viene scelto il primo addendo.
somma = 0                    # Il totale iniziale è zero.
while numero <= 4:           # Vengono inclusi gli addendi da 1 a 4.
    somma = somma + numero    # Il valore corrente viene aggiunto al totale.
    numero = numero + 1      # Viene preparato l'addendo successivo.
print("Somma:", somma)        # Viene mostrato il totale finale.
````

### Come cambia l'accumulatore
| Numero | Somma prima | Calcolo | Somma dopo |
|---|---|---|---|
| 1 | 0 | 0 + 1 | 1 |
| 2 | 1 | 1 + 2 | 3 |
| 3 | 3 | 3 + 3 | 6 |
| 4 | 6 | 6 + 4 | 10 |

In `somma = somma + numero` viene prima calcolato il lato destro; poi il nome `somma` viene associato al risultato.

### Gli errori principali nel `while`
Un aggiornamento assente, esterno al corpo o nella direzione sbagliata può impedire la terminazione. Con partenza da 1, `< 5` esclude 5 mentre `<= 5` lo include. Se una cella continua a eseguire, l'esecuzione può essere interrotta con il comando di arresto del notebook.


## 4. L’oggetto creato da `range()`

`range()` crea un oggetto di tipo `range`: una sequenza immutabile di interi descritta da inizio, limite escluso e passo. L’oggetto conserva i parametri e calcola i valori quando servono, senza memorizzarli tutti.

In questo esempio, `intervallo` fa riferimento all’oggetto che rappresenta 2, 4, 6, 8 e 10.


````python
intervallo = range(2, 11, 2)  # Vengono stabiliti inizio 2, limite escluso 11 e passo 2.
print(type(intervallo))        # Viene mostrato il tipo dell'oggetto: range.
print(intervallo)              # Viene mostrata la descrizione dell'intervallo.
````

## `range()`: inizio, limite e passo
| Forma | Esempio | Valori rappresentati |
|---|---|---|
| `range(fine)` | `range(4)` | 0, 1, 2, 3 |
| `range(inizio, fine)` | `range(1, 5)` | 1, 2, 3, 4 |
| `range(inizio, fine, passo)` | `range(2, 9, 2)` | 2, 4, 6, 8 |

L'inizio predefinito è 0 e il passo predefinito è 1. **Il limite finale è sempre escluso.** Gli argomenti sono interi e il passo deve essere diverso da zero. Con passo positivo i valori rimangono minori del limite; con passo negativo rimangono maggiori del limite.


## 5. Creare e osservare un intervallo

La stampa dell’oggetto ne mostra la descrizione; `len()` restituisce il numero di valori rappresentati. In questo caso sono cinque. `range()` descrive i valori; il ciclo `for` permette di percorrerli.


````python
intervallo = range(2, 11, 2)  # L'intervallo rappresenta 2, 4, 6, 8 e 10.
print(intervallo)              # Viene mostrato range(2, 11, 2).
print(len(intervallo))         # Viene mostrato il numero di valori: 5.
````

### La scelta dei valori da percorrere
| Obiettivo | Espressione | Valori |
|---|---|---|
| Cinque valori a partire da zero | `range(5)` | 0, 1, 2, 3, 4 |
| Da 1 a 5 inclusi | `range(1, 6)` | 1, 2, 3, 4, 5 |
| Dispari da 1 a 9 | `range(1, 10, 2)` | 1, 3, 5, 7, 9 |
| Da 5 a 1 in discesa | `range(5, 0, -1)` | 5, 4, 3, 2, 1 |
| Passo incompatibile con i limiti | `range(5, 0)` | Nessun valore |

## `for`: lo schema concettuale
Lo schema riprende il costrutto iniziale e contiene segnaposto da sostituire con le parti del programma.

```text
for nome in intervallo:      # Il nome assume il valore successivo dell'intervallo.
    istruzioni del corpo    # Il blocco viene eseguito per ciascun valore.
istruzioni successive       # L'esecuzione prosegue dopo l'ultimo valore.
```


## 6. `for`: contare da 1 a 5

Il nome `numero` assume i valori rappresentati da `range(1, 6)`. Il passaggio al valore successivo è automatico.

**Nota bene.** L’indentazione delimita il corpo del `for`: solo `print(numero)` viene ripetuto. La stampa di `Fine` avviene dopo il ciclo.


````python
for numero in range(1, 6):  # Vengono percorsi gli interi da 1 a 5 inclusi.
    print(numero)           # Viene mostrato un valore a ogni iterazione.
print("Fine")                # Il messaggio viene mostrato dopo l'ultimo valore.
````

## 7. `for`: passo positivo

Il passo 2 permette di ottenere i pari da 2 a 10; il limite 11 è escluso.


````python
for numero in range(2, 11, 2):  # Vengono forniti 2, 4, 6, 8 e 10.
    print(numero)               # Viene mostrato il numero pari corrente.
````

## 8. `for`: passo negativo

Il passo −1 permette di scendere da 5 a 1; lo zero è escluso.


````python
for numero in range(5, 0, -1):  # Vengono forniti 5, 4, 3, 2 e 1.
    print(numero)               # Viene mostrato il valore corrente.
````

## 9. La tabellina di un numero

`n` rimane uguale a 3; `fattore` cambia da 1 a 5. Ogni iterazione produce una riga della tabellina. La chiamata a `print()` continua sulla riga successiva grazie alle parentesi aperte.


````python
n = 3                            # Viene fissato il numero della tabellina.
for fattore in range(1, 6):      # Il fattore assume i valori da 1 a 5.
    prodotto = n * fattore       # Viene calcolato il prodotto corrente.
    print(n, "x", fattore,       # Viene iniziata la stessa istruzione di stampa.
          "=", prodotto)         # L'istruzione viene completata con il risultato.
````

## 10. Sommare da 1 a n con `for`

Viene acquisito un intero positivo. Il limite `n + 1` permette di includere `n`. Con input 4 si ottiene `Somma: 10`; la stampa esterna mostra solo il totale finale.


````python
n = int(input("Limite: "))      # Viene letto e convertito l'ultimo addendo.
somma = 0                       # Viene inizializzato il totale.
for numero in range(1, n + 1):  # Vengono percorsi tutti gli addendi da 1 a n.
    somma = somma + numero       # L'addendo corrente viene aggiunto al totale.
print("Somma:", somma)           # Viene mostrata la somma al termine del ciclo.
````

## 11. Lo stesso risultato: versione con `while`

La stampa di 1, 2 e 3 richiede un’inizializzazione, una condizione e un aggiornamento espliciti.


````python
numero = 1                 # Viene inizializzato il contatore.
while numero <= 3:         # Viene incluso il valore 3.
    print(numero)          # Viene mostrato il valore corrente.
    numero = numero + 1    # Il contatore viene incrementato.
````

## 12. Lo stesso risultato: versione con `for`

Il medesimo risultato si ottiene definendo l’intervallo con `range()`. Il ciclo assegna automaticamente i valori a `numero`.


````python
for numero in range(1, 4):  # Vengono forniti 1, 2 e 3.
    print(numero)           # Viene mostrato ciascun valore.
````

## 13. Limiti e sequenze vuote

`range(1, 1)` è vuoto perché inizio e limite escluso coincidono. `range(3, 0)` è vuoto perché il passo predefinito +1 non permette di scendere. I due corpi vengono saltati; compaiono solo i tre messaggi. Per ottenere 3, 2, 1 si userebbe `range(3, 0, -1)`.


````python
print("Primo ciclo")          # Viene introdotto il primo ciclo.
for numero in range(1, 1):   # L'intervallo non contiene valori.
    print(numero)            # Questa istruzione non viene eseguita.
print("Secondo ciclo")        # Viene introdotto il secondo ciclo.
for numero in range(3, 0):   # Il passo positivo rende vuoto l'intervallo.
    print(numero)            # Anche questa istruzione viene saltata.
print("Fine")                 # L'esecuzione prosegue dopo i due cicli.
````


## Esercizio svolto con `while` · La somma di n numeri
**Consegna.** Si richiede un programma che legga una quantità n compresa tra 1 e 10, acquisisca esattamente n numeri interi e ne mostri la somma. È richiesto un ciclo `while`.

**Caso di prova.** Con n = 3 e valori 4, −2, 5, il risultato atteso è `Somma: 7`.


````python
n = int(input("Quantita da 1 a 10: "))  # Viene acquisito il numero di dati da leggere.
letti = 0                               # All'inizio non è stato acquisito alcun dato.
somma = 0                               # Il totale viene inizializzato a zero.
while letti < n:                       # Il corpo si ripete finché mancano dati.
    valore = int(input("Numero: "))    # Viene acquisito un nuovo intero a ogni passaggio.
    somma = somma + valore              # Il dato letto viene aggiunto al totale.
    letti = letti + 1                   # Viene conteggiata la lettura appena effettuata.
print("Somma:", somma)                  # Viene mostrato il totale dopo n letture.
````

**Spiegazione della soluzione.** `letti` conta gli inserimenti e `somma` accumula i valori acquisiti. La lettura si trova dentro il ciclo perché serve un nuovo dato a ogni iterazione. Dopo il terzo inserimento, `letti` vale 3 e la condizione è falsa. La stampa esterna mostra soltanto il totale finale.

## Esercizio svolto con `for` · Il fattoriale
**Consegna.** Si richiede un programma che acquisisca un intero n compreso tra 0 e 10 e calcoli n! mediante un ciclo `for`. Il fattoriale è il prodotto degli interi da 1 a n; per definizione, 0! = 1.

**Casi di prova.** Con n = 4 il risultato è 24; con n = 0 il risultato è 1.


````python
n = int(input("Intero da 0 a 10: "))  # Viene acquisito il numero di cui calcolare il fattoriale.
fattoriale = 1                        # Il prodotto parte dall'elemento neutro 1.
for fattore in range(1, n + 1):      # Vengono forniti i fattori da 1 a n inclusi.
    fattoriale = fattoriale * fattore # Il prodotto viene aggiornato con il fattore corrente.
print("Fattoriale:", fattoriale)      # Viene mostrato il prodotto finale.
````

**Spiegazione della soluzione.** L'accumulatore parte da 1 perché un prodotto inizializzato a zero rimarrebbe zero. Con n = 4, i prodotti parziali sono 1, 2, 6 e 24. Con n = 0, l'intervallo `range(1, 1)` è vuoto e l'accumulatore rimane 1.

**Consegna.** Si richiede di leggere un intero n e di stampare tutti gli interi da 1 a n, estremi inclusi, uno per riga.

````python
# Codice per l'esercizio 01

n = int(input("Inserisci un numero intero: "))  # Viene acquisito un numero intero.

i = 1  # Viene inizializzato il contatore per la tabellina.

while i <= n:  # Il ciclo continua finché il contatore non supera 10.
    print(f"{i}")
    i = i + 1  # Il contatore viene incrementato per la prossima iterazione.

print("Fine conteggio")  # Viene mostrato un messaggio di fine esercizio.


# CON IL CICLO FOR

for i in range(1, n + 1):  # Vengono forniti i valori da 1 a n inclusi.
    print(f"{i}")  # Viene mostrato il valore corrente del contatore.

print("Fine conteggio")  # Viene mostrato un messaggio di fine esercizio.
````

## Esercizio 2 · Conto alla rovescia

**Consegna.** Si richiede di leggere un intero n e di stampare i numeri da n a 1, includendo entrambi gli estremi.

**Richiesta.** È richiesto un ciclo `while`. Dopo lo zero deve comparire `Partenza!`.

**Casi di prova.** Con n = 3 devono comparire 3, 2, 1 e il messaggio. Con n = 1 devono comparire 1 e il messaggio.


````python
# Codice per l'esercizio 02

n = int(input("Inserisci un numero intero: "))  # Viene acquisito un numero intero.

i = n  # Viene inizializzato il contatore per il conto alla rovescia.

while i > 0:  # Il ciclo continua finché il contatore non scende sotto 1.
    print(f"{i}")
    i = i - 1  # Il contatore viene decrementato per la prossima iterazione.

print("Partenza!")  # Viene mostrato un messaggio di partenza dopo il conto alla rovescia.

print(type(range(n, 0, -1)))
print(range(n, 0, -1))

# CON IL CICLO FOR
for i in range(n, 0, -1):  # Il ciclo for genera i numeri da n a 1.
    print(type(i))  # Viene mostrato il numero corrente.
    print(f"{i}")  # Viene mostrato il numero corrente.

print("Partenza!")  # Viene mostrato un messaggio di partenza dopo il conto alla rovescia.
````
