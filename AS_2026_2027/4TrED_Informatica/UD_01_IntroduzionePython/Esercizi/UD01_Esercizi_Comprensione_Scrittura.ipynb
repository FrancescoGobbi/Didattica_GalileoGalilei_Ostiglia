### A4 · Traccia di un while

**Quesito.** Qual è l’output del seguente codice?


````python
numero = 2
somma = 0
while numero <= 8:
    somma = somma + numero
    numero = numero + 2
print("Somma:", somma)
print("Numero:", numero)
````

### A5 · Contatore e accumulatore

**Quesito.** Qual è l’output del seguente codice?


````python
conteggio = 0
somma = 0
for numero in range(1, 8):
    if numero % 3 == 0:
        conteggio = conteggio + 1
        somma = somma + numero
print(conteggio, somma)
````
