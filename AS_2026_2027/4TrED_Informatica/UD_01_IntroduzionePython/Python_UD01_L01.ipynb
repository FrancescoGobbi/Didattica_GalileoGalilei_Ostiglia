## Prima del codice · Python e i suoi oggetti

Python è un linguaggio di programmazione ad alto livello: scriviamo istruzioni che un interprete esegue. Un notebook alterna celle di testo e celle di codice; i nomi definiti rimangono disponibili nella sessione.

| Concetto | Significato | Esempio concreto |
|---|---|---|
| Classe | Definisce un tipo di oggetti e il loro comportamento | Bicicletta |
| Oggetto | Una specifica istanza di una classe | La bicicletta di Anna |
| Attributo | Un dato associato a un oggetto | Il suo colore o la sua velocità |
| Metodo | Un'operazione associata a una classe, richiamabile su un oggetto | `frena()` |

L'esempio della bicicletta è concettuale: non definiamo ancora nuove classi. In Python utilizzeremo oggetti già disponibili, come numeri e stringhe. `print()` e `input()` sono funzioni; `nome.upper()` è una chiamata a un metodo.


## Le prime righe: nomi, valori e commenti

`#` introduce un commento. `=` associa un nome a un oggetto: qui una stringa e un intero. `print()` mostra il risultato.

**Risultato:** `Ciao Anna`, poi `17`. L'espressione `eta + 1` non modifica il valore associato a `eta`.


````python
# Un commento spiega il codice
nome = "Anna"
eta = 16
print("Ciao", nome)
print(eta + 1)
````

## Output: mostrare testo e calcoli con `print()`

Gli argomenti separati da virgole sono stampati con uno spazio tra loro; ogni chiamata termina normalmente andando a capo.

**Risultato:** `Ciao Anna`, `Punti: 12`, `5`, `2 + 3`, su quattro righe. Senza virgolette Python calcola l'espressione; tra virgolette conserva il testo.


````python
nome = "Anna"
punti = 12
print("Ciao", nome)
print("Punti:", punti)
print(2 + 3)
print("2 + 3")
````

### Riconoscere `int`, `float`, `bool` e `str` 

`type()` restituisce il tipo di un oggetto. `12` è un intero; `"12"` è una stringa. Nei numeri decimali Python usa il punto. I valori booleani sono `True` e `False`, con iniziale maiuscola.


````python
print(12, type(12))          # Intero: int.
print(-3, type(-3))          # Anche gli interi negativi sono int.
print(0, type(0))            # Zero è un int.
print(3.5, type(3.5))        # Numero in virgola mobile: float.
print(-0.2, type(-0.2))      # Un float negativo.
print(True, type(True))      # Valore booleano: bool.
print(False, type(False))    # L'altro valore booleano.
print("ciao", type("ciao"))  # Testo: str.
print("12", type("12"))      # Le cifre tra virgolette sono testo.
````

## Convertire l’input: testo e numero non sono la stessa cosa

`int(testo)` converte una stringa che rappresenta un intero. Non modifica `testo`: il risultato è associato a `eta`.

**Prova:** inserisci `16`. `eta + 1` vale `17`; `testo + "1"` concatena due stringhe e produce `161`. Si assume un input valido: `int("ciao")` genera `ValueError`.


````python
testo = input("Eta: ")
eta = int(testo)
print(eta + 1)
print(testo + "1")
````

### Convertire un numero decimale

`float("2.5")` produce il numero `2.5`. Inserisci il punto, non la virgola: `"2,5"` non è una rappresentazione accettata da `float()`.


````python
numero = float("2.5")  # Converte il testo in un float.
print(numero)           # Mostra 2.5.
print(type(numero))     # Conferma il tipo float.
````

## Un metodo delle stringhe: `upper()`

`nome.upper()` restituisce una stringa con le lettere maiuscole. Le stringhe sono **immutabili**: non si modifica l'oggetto originale.

**Risultato:** `ANNA`, poi `Anna`. Il risultato del metodo è associato a `maiuscolo`; `nome` conserva il riferimento alla stringa iniziale.


````python
nome = "Anna"
maiuscolo = nome.upper()
print(maiuscolo)
print(nome)
````

### Calcolare con i numeri

`/` esegue la divisione e restituisce un `float`; `//` arrotonda il quoziente verso il basso; `%` restituisce il resto; `**` esegue una potenza.
Moltiplicazione e divisione precedono somma e sottrazione. Le parentesi cambiano l'ordine dei calcoli.


````python
print(7 + 2)       # Somma: 9.
print(7 - 2)       # Differenza: 5.
print(7 * 2)       # Prodotto: 14.
print(7 / 2)       # Divisione: 3.5.
print(7 // 2)      # Quoziente arrotondato verso il basso: 3.
print(7 % 2)       # Resto: 1.
print(2 ** 3)      # Potenza: 8.
print(2 + 3 * 4)   # Prima 3 * 4: risultato 14.
print((2 + 3) * 4) # Prima 2 + 3: risultato 20.
print(6 / 2)       # Anche qui il risultato è float: 3.0.
print(-7 // 2)     # -3.5 arrotondato verso il basso: -4.
````

## Il risultato di un’operazione è un oggetto

`a + b` produce il valore `13`, che associamo al nome `risultato`. Gli oggetti numerici sono immutabili: `a` e `b` continuano a riferirsi a `10` e `3`.

Il risultato è un oggetto distinto dagli operandi in questo esempio. **Non significa che ogni operazione allochi sempre un nuovo oggetto in memoria:** Python può riutilizzare oggetti immutabili. Le stampe aggiunte rendono visibili i valori.


````python
a = 10
b = 3
risultato = a + b
print("Risultato:", risultato)  # 13.
print("Operandi:", a, b)        # 10 e 3, invariati.
````

## Tre etichette, un solo oggetto

L'assegnazione `b = a` associa `b` allo stesso oggetto già associato ad `a`; non ne crea automaticamente una copia.

```text
a ──┐
b ──┼──► oggetto int con valore 10
c ──┘
```

**Risultato:** `10 10 10`. Un nome è un riferimento a un oggetto; non è un puntatore su cui effettuare operazioni aritmetiche.


````python
a = 10
b = a
c = a
print(a, b, c)
````

## Riassegnare un nome lascia invariati gli altri riferimenti

Python prima valuta `a + 1`, poi associa `a` al risultato. Il nome `b` resta associato al precedente oggetto.

```text
Prima:  a ──┐                 Dopo:  a ──► int 11
        b ──┴──► int 10              b ──► int 10
```

**Risultato:** `11 10`. Non è stato modificato l'intero `10`: è cambiata l'associazione del nome `a`.


````python
a = 10
b = a
a = a + 1
print(a, b)  # 11 10
````

### Riferimenti e garbage collector

Questa cella traduce lo schema delle slide in codice. `object()` crea un semplice oggetto dimostrativo e `a` e `b` lo indicano entrambi. `del a` elimina il nome `a`, non il nome `b` né necessariamente l'oggetto.

Dopo `del b`, se non restano altri riferimenti, l'oggetto può essere recuperato dalla gestione automatica della memoria. Non stabiliamo quando ciò avvenga: implementazioni Python e notebook possono gestire o trattenere riferimenti in modi diversi.


````python
a = object()                 # Crea l'oggetto dimostrativo.
b = a                        # Un secondo nome indica lo stesso oggetto.
print(a is b)                # True: identità dell'oggetto, non confronto di valori.
del a                        # Rimuove solo il nome a.
print(type(b))               # b permette ancora di raggiungere l'oggetto.
del b                        # Rimuove anche il secondo nome.
print("Entrambi i nomi sono stati eliminati.")
````

## Confrontare valori per ottenere una condizione

Un confronto restituisce `True` o `False`. `==` confronta valori; `=` esegue un'assegnazione.

**Risultato:** `False`, poi `True`: 16 non è almeno 18, ma è uguale a 16.


````python
eta = 16
print(eta >= 18)
print(eta == 16)
````

### Tutti gli operatori di confronto 

`!=` significa diverso; `<=` e `>=` includono l'uguaglianza. Questi confronti rendono eseguibile la tabella della slide.


````python
print(16 == 16)  # True: uguali.
print(16 != 18)  # True: diversi.
print(16 < 18)   # True: minore.
print(16 > 18)   # False: non maggiore.
print(18 <= 18)  # True: minore o uguale.
print(18 >= 18)  # True: maggiore o uguale.
````

## Solo `if`: un blocco facoltativo

Il corpo indentato si esegue solo se la condizione è vera. I due punti introducono il blocco; usa quattro spazi di indentazione.

Con `32` compaiono `Fa caldo` e `Fine controllo`. L'ultima stampa è esterna al blocco e si esegue sempre.


````python
temperatura = 32
if temperatura > 30:
    print("Fa caldo")
print("Fine controllo")
````

### Lo stesso `if` con condizione falsa

Con `20` il corpo viene saltato: compare solo `Fine controllo`.


````python
temperatura = 20
if temperatura > 30:
    print("Fa caldo")
print("Fine controllo")
````

## `if` ed `else`: scegliere tra due alternative

Si esegue esattamente uno dei due blocchi. `else` non ha una condizione: raccoglie il caso in cui quella di `if` è falsa.

Con `16` il risultato è `Minorenne`.


````python
eta = 16
if eta >= 18:
    print("Maggiorenne")
else:
    print("Minorenne")
````

### La seconda prova con età 18

Con `18` il confronto è vero: si stampa `Maggiorenne`.


````python
eta = 18
if eta >= 18:
    print("Maggiorenne")
else:
    print("Minorenne")
````

### La seconda prova con età 21

Con `21` il confronto è vero: si stampa `Maggiorenne`.


````python
eta = 21
if eta >= 18:
    print("Maggiorenne")
else:
    print("Minorenne")
````

## `if`, `elif`, `else`: più alternative in ordine

Python valuta le condizioni dall'alto e si ferma al primo ramo vero. `else` raccoglie i casi restanti.

Con `7`, il primo confronto è falso e il secondo è vero: compare `Sufficiente`. Le soglie vanno dalla più alta alla più bassa.


````python
voto = 7
if voto >= 8:
    print("Ottimo")
elif voto >= 6:
    print("Sufficiente")
else:
    print("Da migliorare")
````

### Combinare condizioni: `and`, `or`, `not`

Con condizioni booleane, `and` richiede che entrambe siano vere, `or` richiede che almeno una sia vera e `not` nega la condizione.

| A | B | A and B | A or B |
|---|---|---|---|
| True | True | True | True |
| True | False | False | True |
| False | True | False | True |
| False | False | False | False |

La precedenza è `not`, poi `and`, poi `or`. Le parentesi rendono esplicito il ragionamento. In generale `and` e `or` restituiscono uno degli operandi; qui usiamo esclusivamente valori booleani.


````python
print(True and True, True or True)     # True True.
print(True and False, True or False)   # False True.
print(False and True, False or True)   # False True.
print(False and False, False or False) # False False.
print(not True)                       # False.
print(not False)                      # True.
````

## Unire i controlli: il permesso di accesso

L'accesso richiede due cose: essere maggiorenne **oppure** avere un permesso, **e** non essere sospeso.

`eta >= 18` è `False`; `ha_permesso` è `True`, quindi la parentesi vale `True`. Anche `not sospeso` vale `True`: l'accesso è consentito e la stampa mostra `True`.


````python
eta = 16
ha_permesso = True
sospeso = False
accesso = (eta >= 18 or ha_permesso) and not sospeso
print(accesso)
````

### Accesso negato in caso di sospensione

Con `sospeso = True`, `not sospeso` diventa `False`: il risultato finale è `False`, anche se esiste un permesso.


````python
eta = 16
ha_permesso = True
sospeso = True
accesso = (eta >= 18 or ha_permesso) and not sospeso
print(accesso)
````
